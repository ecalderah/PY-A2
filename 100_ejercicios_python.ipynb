### Las cinco reglas que hacen que esto funcione

5. **La verificación es un espejo, no un juez.** Las celdas de verificación traen los valores esperados a la vista, así que siempre se pueden "aprobar" devolviéndolos a mano. Hacerlo solo te engaña a ti: el ✅ no mide si tu función sirve, mide si coincidió en esos casos. Si dudas, cámbiale los datos de entrada al ejercicio y vuelve a correrlo — una solución que solo funciona con los números de la prueba todavía no está terminada.

| ⏳ | todavía no defines la función o variable — en los 🔁 significa que te falta un ejercicio anterior |
| 💥 | tu código lanzó una excepción; el nombre del error te dice qué revisar |

## Módulo 0 — Antes de empezar: cómo leer un error

Esto va primero por una razón: **vas a ver muchos errores**, y no son señal de que algo esté mal contigo. Un error de Python es un informe técnico bastante generoso, si sabes por dónde leerlo.

Cuando algo falla, Python imprime un **traceback** (*rastreo*). Se lee **de abajo hacia arriba**:

```
Traceback (most recent call last):
  File "<ipython-input-3>", line 5, in <module>
    resultado = dividir(10, 0)
  File "<ipython-input-3>", line 2, in dividir
    return a / b
ZeroDivisionError: division by zero
```

1. **La última línea es la respuesta**: el **tipo** de error (`ZeroDivisionError`) y el **mensaje** (`division by zero`).
2. **La penúltima te dice dónde**: archivo, número de línea y la línea de código exacta.
3. **Las de arriba son el camino**: qué llamó a qué hasta llegar ahí. Lee de abajo hacia arriba hasta encontrar la primera línea que sea *tu* código.

### Los ocho errores que verás una y otra vez

| Error | Qué significa realmente | Primera cosa que revisar |
|---|---|---|
| `SyntaxError` | el código no es Python válido (se detecta antes de correr) | paréntesis sin cerrar, dos puntos faltantes, comillas desparejas |
| `IndentationError` | la sangría no cuadra | mezcla de espacios y tabuladores |
| `NameError` | ese nombre no existe | un typo, o una celda anterior que no corriste |
| `TypeError` | operación entre tipos incompatibles | ¿estás sumando texto con número? ¿olvidaste un `return`? |
| `ValueError` | tipo correcto, valor imposible | `int("abc")`, `math.sqrt(-1)` |
| `IndexError` | posición fuera de rango | ¿cuenta desde 0? ¿la lista está vacía? |
| `KeyError` | llave inexistente en un diccionario | revisa mayúsculas y espacios en la llave |
| `AttributeError` | el objeto no tiene ese método | ¿es una tupla y no una lista? ¿es `None`? |

**El caso especial más frecuente en Colab:** `NameError` cuando *sí* definiste la variable. Casi siempre es porque corriste las celdas fuera de orden o reiniciaste el entorno. Solución: `Entorno de ejecución → Reiniciar y ejecutar todo`.


````python
# Corre esta celda: provoca un error a PROPÓSITO para que veas un traceback real.
# No la borres — vuelve a ella cuando algo falle y no entiendas el mensaje.

def dividir(a, b):
    return a / b

resultado = dividir(10, 0)   # <- ZeroDivisionError. Lee el traceback de abajo hacia arriba.
print(resultado)             # esta línea nunca se ejecuta
````

### Corre esta celda para que funcione el proyecto.

Corre la celda de abajo **antes que cualquier otra**. Define el verificador de ejercicios (`probar`), la utilidad `tipo_de_error` que usan los ejercicios del Módulo 9, y `mi_progreso()` para ver tu avance. Si reinicias el entorno, vuelve a correrla.

````python
# ============================================================================
#  CELDA DE ARRANQUE — córrela UNA VEZ antes de empezar (y de nuevo si
#  reinicias el entorno). Define el verificador de ejercicios y las utilidades
#  del curso. No necesitas entender este código todavía.
# ============================================================================
import math, os, io, json, traceback

try:
    import numpy as np
except ImportError:
    np = None
try:
    import pandas as pd
except ImportError:
    pd = None

_PROGRESO = {}


def _mostrar(v):
    """repr() legible: evita el ruido de np.float64(...) en los mensajes."""
    try:
        if np is not None and isinstance(v, np.generic):
            return repr(v.item())
    except Exception:
        pass
    return repr(v)


def _iguales(a, b, tol=1e-9):
    """Compara resultados tolerando ruido de punto flotante y tipos de NumPy/pandas."""
    if isinstance(b, bool) or isinstance(a, bool):
        return a is b
    if b is None:
        return a is None
    if isinstance(b, dict):
        if not isinstance(a, dict) or set(a.keys()) != set(b.keys()):
            return False
        return all(_iguales(a[k], b[k], tol) for k in b)
    if isinstance(b, (list, tuple)):
        if not isinstance(a, type(b)) or len(a) != len(b):
            return False
        return all(_iguales(x, y, tol) for x, y in zip(a, b))
    if isinstance(b, (int, float)) and isinstance(a, (int, float)):
        return abs(a - b) <= tol * max(1.0, abs(b))
    if np is not None and (hasattr(b, "shape") or hasattr(a, "shape")):
        try:
            return bool(np.allclose(np.asarray(a, dtype=float),
                                    np.asarray(b, dtype=float), rtol=1e-7, atol=1e-9))
        except Exception:
            pass
    return a == b


def probar(titulo, casos, tol=1e-9):
    """Corre los casos de prueba de un ejercicio e imprime el resultado.

    `casos` es una lista de pares (funcion_sin_argumentos, valor_esperado).
    Las funciones se escriben como `lambda: ...` por una razón práctica: así
    cada caso se ejecuta por separado y un error en uno no tumba los demás.
    Verás `lambda` explicado formalmente en el módulo de estilo pythonic; aquí
    solo hace de envoltura.
    """
    ok = 0
    lineas = []
    for i, (obtener, esperado) in enumerate(casos, start=1):
        try:
            obtenido = obtener()
        except NameError as e:
            # Un NameError puede ser dos cosas muy distintas: que todavia no
            # exista la funcion del ejercicio (⏳), o que exista y tenga un
            # typo adentro (💥). La profundidad del traceback las separa.
            hondo = len(traceback.extract_tb(e.__traceback__)) > 2
            if hondo:
                lineas.append(f"  💥 caso {i}: NameError dentro de tu código — {e}")
            else:
                lineas.append(f"  ⏳ caso {i}: falta definir algo ({e})")
            continue
        except Exception as e:
            lineas.append(f"  💥 caso {i}: {type(e).__name__}: {e}")
            continue
        if _iguales(obtenido, esperado, tol):
            ok += 1
            lineas.append(f"  ✅ caso {i}: {_mostrar(obtenido)}")
        else:
            aviso = ""
            if type(obtenido) is not type(esperado):
                aviso = (f"   ← ojo: devolviste un {type(obtenido).__name__} "
                         f"y se esperaba un {type(esperado).__name__}")
            lineas.append(f"  ❌ caso {i}: obtuviste {_mostrar(obtenido)} — "
                          f"se esperaba {_mostrar(esperado)}{aviso}")

    total = len(casos)
    _PROGRESO[titulo] = (ok, total)
    encabezado = "🎉 ¡Correcto!" if ok == total else "🔧 Aún no — revisa los casos marcados"
    print(f"{titulo}: {ok}/{total} — {encabezado}")
    print("\n".join(lineas))


def tipo_de_error(f, *args, **kwargs):
    """Devuelve el NOMBRE de la excepción que lanza f(*args), o 'sin error'."""
    try:
        f(*args, **kwargs)
    except Exception as e:
        return type(e).__name__
    return "sin error"


def mi_progreso(detalle=False):
    """Resumen de los ejercicios verificados en esta sesión."""
    if not _PROGRESO:
        print("Todavía no has corrido ninguna celda de verificación.")
        return

    def clave(t):
        try:
            return int(t.split()[-1])
        except ValueError:
            return 0

    completos = [t for t, (o, n) in _PROGRESO.items() if o == n]
    parciales = [t for t, (o, n) in _PROGRESO.items() if o != n]
    print(f"Ejercicios intentados : {len(_PROGRESO)}")
    print(f"Resueltos al 100%     : {len(completos)}")
    print(f"Pendientes            : {len(parciales)}")
    ancho = 40
    hechos = int(ancho * len(completos) / max(1, len(_PROGRESO)))
    print("[" + "█" * hechos + "░" * (ancho - hechos) + "]")
    if parciales:
        print("\nPor revisar: " + ", ".join(sorted(parciales, key=clave)))
    if detalle:
        for t in sorted(_PROGRESO, key=clave):
            o, n = _PROGRESO[t]
            print(f"  {'✅' if o == n else '❌'} {t}: {o}/{n}")


print("Entorno listo ✅")
print(f"numpy      : {np.__version__ if np is not None else 'no disponible'}")
print(f"pandas     : {pd.__version__ if pd is not None else 'no disponible'}")
print("Utilidades : probar(), tipo_de_error(), mi_progreso()")
````

### Teoría

Un programa hace tres cosas: **guarda valores**, **los transforma** y **muestra resultados**. Este módulo cubre las tres en su forma más simple.

Una **variable** (*variable*) es un nombre pegado a un valor. El signo `=` no es la igualdad de las matemáticas: es una instrucción que se lee de derecha a izquierda — *"evalúa lo de la derecha y guárdalo con el nombre de la izquierda"*. Por eso `x = x + 1` tiene todo el sentido en Python y ninguno en álgebra.

Todo valor tiene un **tipo** (*type*), y el tipo determina qué operaciones son legales:

| Tipo | Nombre completo | Ejemplo | Para qué |
|---|---|---|---|
| `int` | *integer* | `2026` | conteos, años, índices |
| `float` | *floating point* | `4.7` | tasas, montos, mediciones |
| `str` | *string* | `"Guanajuato"` | texto, etiquetas, claves |
| `bool` | *boolean* | `True` | banderas, resultados de comparaciones |

Python es de **tipado dinámico** (*dynamically typed*): no declaras el tipo, lo deduce del valor. Pero el tipo importa muchísimo: `"3" + "4"` da `"34"` y `3 + 4` da `7`, sin que nadie te avise de que quisiste decir otra cosa.

**Reglas para nombrar variables:** solo letras, dígitos y guiones bajos; no puede empezar con dígito; no puede ser una palabra reservada (`if`, `for`, `class`, ...). Y una regla no obligatoria pero crítica: **no uses nombres de funciones nativas** (`str`, `list`, `sum`, `type`, `input`, `max`). Python te dejará hacerlo y romperá tu sesión 200 líneas después.

````python
# ---- Mini-lección ejecutable: corre esta celda y lee cada salida ----

pib_gto = 1_040_000        # el guion bajo es separador visual de miles: Python lo ignora
inflacion = 4.7
estado = "Guanajuato"
es_estatal = True

print(pib_gto, type(pib_gto))
print(inflacion, type(inflacion))
print(estado, type(estado))
print(es_estatal, type(es_estatal))

# Aritmética: los 7 operadores que existen
print("suma          ", 7 + 3)
print("resta         ", 7 - 3)
print("producto      ", 7 * 3)
print("división real ", 7 / 3)     # SIEMPRE devuelve float, aunque la división sea exacta
print("división ent. ", 7 // 3)    # cociente entero
print("módulo        ", 7 % 3)     # residuo
print("potencia      ", 7 ** 3)    # NO se usa ^

# Reasignación: la etiqueta se mueve a otro valor
contador = 0
contador = contador + 1
contador += 1                      # forma corta, idéntica a la anterior
print("contador =", contador)
````

#### Ejercicio 4 — Convertir entre tipos

**Concepto:** Conversión explícita — *type casting*

Partiendo de `texto_numero = "42"`, construye:

- `numero` → el `42` como entero
- `truncado` → `int(3.9)` (¿redondea o trunca? averígualo)

```python
predicciones = [
    3.5,       # 1. 7 / 2            -> la división real SIEMPRE devuelve float
    3,         # 2. 7 // 2           -> división entera
    -4,        # 3. -7 // 2          -> ¡no es -3! redondea hacia ABAJO, no hacia cero
    1,         # 4. 7 % 2            -> residuo
    512,       # 5. 2 ** 3 ** 2      -> ¡no es 64! ** se agrupa a la DERECHA: 2**(3**2) = 2**9
    "float",   # 6. type(4 / 2)      -> 2.0, no 2: sigue siendo float aunque la división sea exacta
    3,         # 7. int(3.9)         -> trunca hacia cero, NO redondea
]
```

**Por qué:** Las tres sorpresas son las tres que causan bugs reales.

1. **`/` siempre devuelve `float`.** En Python 3, `4 / 2` es `2.0`, no `2`. Si necesitas un entero, `//` o `int()`.
2. **`//` redondea hacia abajo (*floor*), no hacia cero.** `-7 // 2` es `-4` porque −4 es el entero inmediatamente **menor** que −3.5. Esto muerde al calcular índices o repartos con negativos.
3. **`**` es asociativo por la derecha**, único entre los operadores aritméticos: `2 ** 3 ** 2` es `2**9 = 512`, no `(2**3)**2 = 64`. Coincide con la notación matemática $2^{3^2}$.

Y una cuarta, más callada: `int(3.9)` da `3` porque **trunca**; para redondear de verdad es `round(3.9)` → `4`.

</details>

`+` en cambio **no** está definido entre `str` e `int`, así que ahí sí truena con `TypeError`. Es una asimetría afortunada: el error ruidoso te avisa, el silencioso no. Por eso la regla es **convertir en la frontera**: en cuanto un número entra a tu programa como texto, conviértelo, y no lo dejes viajar disfrazado.

## Módulo 2 — Texto y formato

**Ejercicios 11–20** · duración aproximada: 45–70 min

**Al terminar este módulo podrás:**

- Construir mensajes con f-strings y formatear números para presentación
- Usar los métodos principales de `str` para limpiar y transformar texto
- Indexar y rebanar cadenas
- Separar y unir texto, que es la mitad del trabajo al leer archivos reales

### Teoría

Casi todo dato del mundo real llega como texto: un CSV, la respuesta de una API, un formulario. Saber manipular `str` no es un tema decorativo, es la puerta de entrada a los datos.

Las **f-strings** (*formatted string literals*) son la forma moderna de construir texto:

```python
f"{variable} y {expresión * 2} con {numero:,.2f}"
```

Todo lo que va dentro de las llaves se evalúa; lo que va después de los dos puntos es el **formato de presentación**:

| Formato | Efecto | Ejemplo |
|---|---|---|
| `:,.0f` | miles, sin decimales | `21,433,226` |
| `:.2f` | 2 decimales fijos | `4.70` |
| `:.1%` | porcentaje (ya multiplica por 100) | `4.7%` |
| `:>10` / `:<10` / `:^10` | alineación en 10 espacios | `      PIB` |
| `:05d` | entero con ceros a la izquierda | `00087` |

El otro concepto central es la **inmutabilidad** (*immutability*): una cadena no se puede modificar. Todos los métodos (`.upper()`, `.strip()`, `.replace()`) **devuelven una cadena nueva**. Si escribes `texto.strip()` sin asignar el resultado, no pasa absolutamente nada.

Para acceder a partes de una cadena se usa **indexado** (`texto[0]`) y **rebanado** (*slicing*, `texto[2:5]`), con la convención `[inicio:fin)`: el inicio se incluye, el fin no.

````python
# ---- Mini-lección: texto ----

clave = "MX-GTO-2026"
print(clave[0], clave[-1])        # primer y último carácter
print(clave[3:6])                 # rebanada: índices 3, 4, 5
print(clave[::-1])                # invertida

sucio = "  INEGI, 2024  "
print(repr(sucio.strip()))        # repr() muestra las comillas: útil para ver espacios
print(sucio.strip().lower().replace(",", ";"))   # los métodos se encadenan

# Formato numérico: la misma cifra, cinco presentaciones
monto = 1234567.891
print(f"{monto}")
print(f"{monto:,.2f}")
print(f"{monto:>15,.0f}")
print(f"{monto/1e6:.2f} millones")
print(f"{0.04687:.2%}")

# split / join: el ciclo de vida de una línea de CSV
linea = "MX,4.7,2024"
campos = linea.split(",")
print(campos)
print(" | ".join(campos))
````

```python

    "PIB",                    # 1. strip() quita espacios de los DOS extremos, no de en medio
    "Pib Nominal",            # 2. title() pone mayúscula en cada palabra
    "2024",                   # 3. desde el índice 3 hasta el final
    2,                        # 4. la 'i' de "inflación" y la de "ción"
    4,                        # 5. "ñ" es UN carácter: n-i-ñ-o
    "1234",                   # 6. sigue siendo TEXTO, no el número 1234
    ["a", "b", "", "c"],      # 7. entre las dos comas hay una cadena vacía: 4 elementos
    "4.7%",                   # 8. el formato % multiplica por 100 y añade el signo
```

## Módulo 3 — Funciones I — empaquetar una idea

**Ejercicios 21–29** · duración aproximada: 50–80 min

**Al terminar este módulo podrás:**

- Definir funciones con `def` y devolver valores con `return`
- Distinguir con claridad `return` de `print`
- Usar parámetros por defecto y argumentos por palabra clave
- Devolver varios valores y componer funciones entre sí
- Leer código ajeno y predecir su salida

### Teoría

Aquí es donde este curso se sale del orden tradicional, y por una buena razón. Casi todos los cursos dejan las funciones para la mitad del temario; nosotros las vemos **antes** de condicionales y ciclos, porque una función es la unidad natural de código que **se puede probar**. De aquí en adelante, cada ejercicio se verifica con varios casos de prueba en lugar de uno solo.

Una **función** (*function*) es una idea empaquetada con nombre:

```python
def nombre(parametro1, parametro2=valor_por_defecto):
    """Docstring: qué hace la función."""
    resultado = ...          # cuerpo, indentado 4 espacios
    return resultado         # entrega el valor a quien la llamó
```

La distinción más importante del módulo:

| | `print(x)` | `return x` |
|---|---|---|
| Qué hace | muestra `x` en pantalla | entrega `x` al código que llamó |
| ¿Se puede reutilizar? | no, el valor se pierde | sí: `y = f(3) + 1` |
| Dentro de otra función | inútil | esencial |

Una función **sin** `return` devuelve `None` de forma implícita. De ahí viene el error `TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'`, que casi siempre significa *"olvidaste el return"*.

**Vocabulario preciso:** los nombres en la definición son **parámetros** (*parameters*); los valores que pasas al llamarla son **argumentos** (*arguments*).

````python
# ---- Mini-lección: funciones ----

def tasa_crecimiento(valor_final, valor_inicial):
    """Tasa de crecimiento simple, en porcentaje."""
    return (valor_final / valor_inicial - 1) * 100

print(tasa_crecimiento(110, 100))          # 10.0 (o casi: mira el punto flotante)
print(f"{tasa_crecimiento(110, 100):.1f}%")

# Parámetro por defecto + llamada por palabra clave
def valor_futuro(capital, tasa=0.085, anios=1):
    """Interés compuesto: M = C (1+i)^n."""
    return capital * (1 + tasa) ** anios

print(valor_futuro(1000))                       # usa los dos defaults
print(valor_futuro(1000, anios=10))             # sobreescribe solo uno
print(valor_futuro(anios=10, capital=1000))     # el orden deja de importar

# return vs print
def suma_print(a, b):
    print(a + b)

def suma_return(a, b):
    return a + b

x = suma_print(2, 3)      # imprime 5...
print("x vale:", x)       # ...pero x es None
y = suma_return(2, 3)
print("y vale:", y, "y y+1 =", y + 1)
````

> 🔮 **Predice antes de correr.** ¿Qué imprime `print(suma_print(2,3) is None)`? Piénsalo antes de correr la celda.

<details>
<summary>🔑 Solución del ejercicio 25 — <em>ábrela solo después de intentarlo</em></summary>

```python
def ordenar_par(a, b):
    """Devuelve (menor, mayor)."""
    if a > b:
        a, b = b, a      # intercambio simultáneo: no hace falta variable temporal
    return a, b          # esto construye una tupla (a, b)

menor, mayor = ordenar_par(9, 3)   # desempaquetado
print(menor, mayor)
```

**Por qué:** Python no devuelve "varios valores": devuelve **una** tupla que luego desempaquetas. El intercambio `a, b = b, a` funciona porque el lado derecho se evalúa completo antes de asignar.

</details>

## Módulo 4 — Decisiones

**Ejercicios 30–40** · duración aproximada: 60–95 min

**Al terminar este módulo podrás:**

- Escribir condiciones con operadores de comparación y lógicos
- Usar `if` / `elif` / `else` y entender por qué el orden importa
- Devolver booleanos sin envolverlos en un `if` redundante
- Traducir una tabla de reglas (un reglamento, una tarifa) a código
- Detectar código muerto por orden de reglas mal puesto

### Teoría

Un programa que solo calcula es una calculadora. Un programa que **decide** empieza a ser un modelo.

Los **operadores de comparación** producen booleanos: `==`, `!=`, `<`, `<=`, `>`, `>=`. El error de dedo más común es escribir `=` (asignación) donde iba `==` (comparación).

Los **operadores lógicos** combinan booleanos: `and`, `or`, `not`. Su precedencia es `not` > `and` > `or`, y hacen **evaluación de cortocircuito** (*short-circuit*): en `A and B`, si `A` es falso, `B` nunca se evalúa. Eso permite escribir `if datos and datos[0] > 0:` sin que reviente con una lista vacía.

La estructura condicional:

```python
if condicion_1:
    ...
elif condicion_2:      # solo se evalúa si la anterior fue FALSA
    ...
else:
    ...
```

Dos ideas que valen más que la sintaxis:

1. **El orden de los `elif` codifica prioridad.** Una cadena de `elif` es una tabla de reglas donde la primera que se cumple gana. Cuando programas un lineamiento o una tarifa, el orden en que escribes las ramas *es* la interpretación que estás adoptando.
2. **Si la condición ya es un booleano, devuélvela.** `return edad >= 18` en lugar de `if edad >= 18: return True else: return False`.

Python también tiene **valores de verdad implícitos** (*truthiness*): `0`, `0.0`, `""`, `[]`, `{}` y `None` son falsos; casi todo lo demás es verdadero. Por eso `if not lista:` significa "si la lista está vacía".

````python
# ---- Mini-lección: decisiones ----

ingreso = 850

if ingreso < 300:
    categoria = "bajo"
elif ingreso < 700:
    categoria = "medio-bajo"
elif ingreso < 1500:
    categoria = "medio-alto"
else:
    categoria = "alto"

print(f"Ingreso {ingreso} -> {categoria}")

# Comparación encadenada: legal y legible en Python
edad = 25
print(18 <= edad <= 65)

# Cortocircuito: la segunda parte NUNCA se evalúa si la primera falla
datos = []
print(len(datos) > 0 and datos[0] > 10)   # False, sin IndexError

# Truthiness: qué es falso en Python
for valor in (0, 0.0, "", [], {}, None, "0", [0], -1):
    print(f"{repr(valor):>6} -> {bool(valor)}")

# El ternario: un if que es una expresión
n = -5
print("positivo" if n > 0 else "no positivo")
````

> 🔮 **Predice antes de correr.** ¿Qué imprime `bool("0")`? ¿Y `bool([])`? Una de las dos suele sorprender.

```python
def dictamen_credito(ingreso, historial):
    """Matriz de decisión 2x2 resuelta con anidamiento."""
    if ingreso >= 10000:
        if historial == "bueno":
            return "aprobado"
        else:
            return "revision"
    else:
        if historial == "bueno":
            return "revision"
        else:
            return "rechazado"

# Alternativa plana, más legible cuando la matriz crece:
def dictamen_plano(ingreso, historial):
    alto = ingreso >= 10000
    bueno = historial == "bueno"
    if alto and bueno:
        return "aprobado"
    if alto or bueno:
        return "revision"
    return "rechazado"
```

## Módulo 5 — Repetición

**Ejercicios 41–55** · duración aproximada: 75–120 min

**Al terminar este módulo podrás:**

- Elegir entre `for` (sabes cuántas vueltas) y `while` (no lo sabes)
- Dominar el patrón acumulador
- Controlar el flujo con `break`, `continue` y el `else` del ciclo
- Recorrer con `enumerate` y `zip` en lugar de `range(len(...))`
- Normalizar acentos y ñ, que es el problema real de trabajar en español
- Diagnosticar un ciclo infinito y escribir ciclos con tope de intentos

### Teoría

La repetición es lo que hace que un programa escale: la misma lógica sobre 10 datos o sobre 10 millones.

**`for`** recorre un iterable: una secuencia conocida de elementos.

```python
for elemento in coleccion:
    ...
```

`range(inicio, fin, paso)` genera enteros y **excluye el fin**: `range(1, 5)` da 1, 2, 3, 4. Esa convención parece incómoda hasta que notas que `len(range(a, b)) == b - a`.

**`while`** repite mientras una condición sea verdadera. Se usa cuando **no sabes de antemano** cuántas iteraciones hacen falta (convergencia, lectura hasta fin de archivo, alcanzar una meta financiera). Todo `while` necesita que el cuerpo mueva la **variable de control** hacia la salida; si no, el ciclo es infinito.

El patrón más importante es el **acumulador**:

```python
total = 0                 # 1. inicializar FUERA
for x in datos:
    total += x            # 2. actualizar DENTRO
return total              # 3. usar DESPUÉS
```

Control fino del flujo:

| Palabra | Efecto |
|---|---|
| `break` | abandona el ciclo inmediatamente |
| `continue` | salta al siguiente elemento |
| `else` (del ciclo) | corre solo si el ciclo terminó **sin** `break` |

Y dos funciones que hacen el código idiomático: `enumerate(seq)` entrega `(índice, valor)`, y `zip(a, b)` recorre dos secuencias en paralelo (truncando con la más corta). Si te encuentras escribiendo `for i in range(len(lista))`, casi siempre querías `enumerate`.

````python
# ---- Mini-lección: ciclos ----

# for con acumulador
total = 0
for i in range(1, 6):
    total += i
print("suma 1..5 =", total)

# while: no sabemos cuántas vueltas
monto, meta, anios = 1000, 2000, 0
while monto < meta:
    monto *= 1.10
    anios += 1
print(f"Duplicar al 10% toma {anios} años (monto final {monto:,.2f})")

# break y continue
for n in range(2, 20):
    if n % 2 == 0:
        continue          # salta los pares
    if n > 10:
        break             # y corta al pasar de 10
    print("impar:", n, end="  ")
print()

# enumerate y zip
paises = ["MX", "BR", "CL"]
inflacion = [4.7, 4.6, 3.9]

for i, p in enumerate(paises):
    print(i, p)

for p, inf in zip(paises, inflacion):
    print(f"{p}: {inf}%")

# for ... else: "terminó sin break"
for x in [1, 3, 5]:
    if x % 2 == 0:
        print("encontré un par")
        break
else:
    print("no había ningún par")
````

> 🔮 **Predice antes de correr.** En el ciclo con `break` y `continue`: ¿cuántos números imprime? Cuéntalos mentalmente antes de correr.

```python
def suma_hasta(n):
    """Suma de los primeros n enteros positivos."""
    total = 0                    # acumulador: se inicializa FUERA del ciclo
    for i in range(1, n + 1):    # range excluye el límite superior, de ahí el +1
        total = total + i        # equivalente: total += i
    return total

# Comprobación con la fórmula de Gauss n(n+1)/2
print(suma_hasta(100), 100 * 101 // 2)
```

**Por qué:** El patrón **acumulador** (inicializar fuera, actualizar dentro, devolver después) es el esqueleto de casi todo cálculo iterativo: sumas, productos, conteos, máximos. Si inicializas dentro del ciclo, se reinicia en cada vuelta y siempre obtendrás el último valor.

```python
import unicodedata

def sin_acentos(texto):
    """Quita marcas diacríticas descomponiendo el texto en Unicode NFD."""
    nfd = unicodedata.normalize("NFD", texto)   # "é" -> "e" + acento combinante
    limpio = ""
    for c in nfd:
        if unicodedata.category(c) != "Mn":     # "Mn" = marca diacrítica
            limpio += c
    return limpio

def contar_vocales_es(texto):
    """Cuenta vocales en español, con o sin acento, sin distinguir mayúsculas."""
    n = 0
    for c in sin_acentos(texto).lower():
        if c in "aeiou":
            n += 1
    return n

# La versión compacta, con comprensión (la verás en el módulo de estilo pythonic):
#   return "".join(c for c in nfd if unicodedata.category(c) != "Mn")

for palabra in ("Querétaro", "MÉXICO", "ñandú", "Tláhuac"):
    print(f"{palabra:>10} -> {sin_acentos(palabra):>10}  vocales: {contar_vocales_es(palabra)}")

# Por qué importa: dos textos que se VEN iguales pueden no serlo
a = "é"                                     # un solo carácter (NFC)
b = unicodedata.normalize("NFD", "é")       # dos caracteres (NFD)
print(a == b, len(a), len(b))               # False 1 2
```

**Por qué:** Unicode permite escribir `é` de dos formas: como un carácter (`U+00E9`, forma NFC) o como `e` más un acento combinante (`U+0065 U+0301`, forma NFD). Se ven idénticas y **no son iguales** para Python. Por eso un `df["estado"] == "Querétaro"` puede devolver cero filas aunque el dato esté ahí: vienen en formas distintas. La normalización antes de comparar es una regla, no una optimización — y en pandas se hace con `df["col"].str.normalize("NFKD")`.

</details>

```python
def cuenta_regresiva(n):
    """Lista de n hasta 1, en orden descendente."""
    resultado = []
    actual = n
    while actual >= 1:          # si n <= 0 nunca entra: devuelve []
        resultado.append(actual)
        actual -= 1             # SIN esta línea el ciclo es infinito
    return resultado
```

```python
def anios_para_meta(capital, tasa, meta):
    """Años completos de interés compuesto para alcanzar una meta.

    Asume tasa > 0: con tasa <= 0 y capital < meta el monto nunca crece
    y el ciclo no terminaría.
    """
    monto = capital
    anios = 0
    while monto < meta:          # con tasa > 0 este ciclo siempre termina
        monto *= (1 + tasa)      # monto = monto * (1 + tasa)
        anios += 1
    return anios

print(anios_para_meta(1000, 0.10, 2000))   # 8
```

**Por qué:** Este es el caso donde `while` es estrictamente mejor que `for`: **no sabes de antemano** cuántas iteraciones hacen falta. La versión analítica sería $n = \lceil \ln(meta/capital)/\ln(1+i) \rceil$; comparar ambas es un buen ejercicio de validación.

```python
def primer_divisor(n):
    """Menor divisor de n mayor que 1 (n mismo si n es primo)."""
    for d in range(2, n + 1):
        if n % d == 0:
            return d      # return sale del ciclo Y de la función de inmediato
    return n              # inalcanzable para n >= 2, pero deja la función total

# Con break sería:
def primer_divisor_break(n):
    encontrado = n
    for d in range(2, n + 1):
        if n % d == 0:
            encontrado = d
            break         # break solo sale del ciclo; la función sigue
    return encontrado
```

**Por qué:** `break` abandona el ciclo más interno; `return` abandona la función completa. Cuando el ciclo está dentro de una función y ya tienes la respuesta, `return` es más limpio: te ahorra la variable bandera.

<summary>🔑 Solución del ejercicio 48 — <em>ábrela solo después de intentarlo</em></summary>

```python
def es_primo(n):
    """True si n es primo. Solo hay que probar divisores hasta la raíz de n."""
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False      # encontró divisor: no es primo
    return True               # el ciclo terminó sin encontrar divisores

# Variante con for/else, que hace explícito "terminó sin break":
def es_primo_else(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            break
    else:
        return True           # este else pertenece al FOR, no a un if
    return False
```

**Por qué:** Basta probar hasta $\sqrt{n}$ porque si $n = a\cdot b$ con $a \le b$, entonces $a \le \sqrt{n}$: cualquier divisor grande viene acompañado de uno pequeño que ya habrías encontrado. El `for...else` es exclusivo de Python y significa "no hubo `break`".

</details>

```python
def tabla_multiplicar(n):
    """Matriz n x n de multiplicaciones."""
    tabla = []
    for i in range(1, n + 1):
        fila = []                   # una fila NUEVA en cada vuelta del ciclo externo
        for j in range(1, n + 1):
            fila.append(i * j)

for fila in tabla_multiplicar(4):
    print(fila)
```

<summary>🔑 Solución del ejercicio 50 — <em>ábrela solo después de intentarlo</em></summary>

```python
def maximo(numeros):
    """Mayor elemento de la lista, o None si está vacía."""
    if not numeros:              # lista vacía es "falsa" en contexto booleano
        return None
    mayor = numeros[0]           # candidato inicial: el primer elemento REAL
    for x in numeros:
        if x > mayor:
            mayor = x
    return mayor
```

**Por qué:** Inicializar con `mayor = 0` es el bug clásico: con `[-5, -2]` devolvería `0`, un valor que no está en los datos. El principio general es inicializar acumuladores con el **elemento neutro correcto**, y para el máximo ese es el primer dato (o `-inf`).

</details>

````python
def indice_del_maximo(numeros):
````

```python
    if not numeros:
```

```python
def producto_punto(a, b):
    """Producto punto (dot product) de dos secuencias numéricas."""
    total = 0
    for x, y in zip(a, b):
        total += x * y
    return total

# Promedio ponderado: pesos que suman 1
print(producto_punto([0.5, 0.3, 0.2], [100, 200, 300]))   # 170.0
```

**Por qué:** `zip` **trunca silenciosamente** a la longitud de la secuencia más corta: si tus listas no tienen el mismo tamaño, no habrá error, solo un resultado incompleto. Para exigir igualdad usa `zip(a, b, strict=True)` (Python 3.10+).

```python
def posicion_mal(objetivo, lista):
```

```python
def posicion(objetivo, lista):
    """Posición 1-based del primer elemento igual a objetivo, o None."""
    i = 0
    while i < len(lista):
        if lista[i] == objetivo:
            return i + 1
        i += 1              # <- LA LÍNEA QUE FALTABA
    return None

print(posicion("b", ["a", "b"]), posicion("z", ["a", "b"]))

# Con for/enumerate el bug es IMPOSIBLE, porque el avance no depende de ti:
def posicion_v2(objetivo, lista):
    for i, x in enumerate(lista, start=1):
        if x == objetivo:
            return i
    return None
```

#### Ejercicio 54 — ➕ Login con límite de intentos

**Concepto:** Ciclo controlado y `input()` — *bounded retry loop*

Un `while True` sin salida es una trampa. Escribe `login(usuario, contrasena, intentos)` donde `intentos` es una lista de tuplas `(usuario_capturado, contrasena_capturada)`, y la regla es:

- si alguno de los **primeros 3** intentos coincide con las credenciales → `"acceso concedido"`
- si se agotan los **3** intentos sin acertar → `"cuenta bloqueada"`
- si la lista tiene **menos de 3** intentos y ninguno acierta → `"acceso denegado"`

Los intentos a partir del cuarto se ignoran (la cuenta ya está bloqueada).

**Por qué recibimos una lista y no usamos `input()`:** para poder probarlo automáticamente. La solución te muestra la versión real con `input()`.

<details>
<summary>💡 Pista</summary>

Recorre con `enumerate` y corta con `break` o `return` en cuanto acierte. Después del ciclo, decide entre bloqueada y denegada según cuántos intentos hubo.

</details>

````python
# ---------- Ejercicio 54: TU SOLUCIÓN ----------
def login(usuario, contrasena, intentos):
    pass
````

````python
# ---------- Verificación (no la edites) ----------
probar("Ejercicio 54", [
    (lambda: login("edgar", "1234", [("edgar", "1234")]), "acceso concedido"),
    (lambda: login("edgar", "1234", [("x", "y"), ("edgar", "1234")]), "acceso concedido"),
    (lambda: login("edgar", "1234", [("a","b"), ("c","d"), ("e","f")]), "cuenta bloqueada"),
    (lambda: login("edgar", "1234", [("a","b"), ("c","d"), ("e","f"), ("edgar","1234")]), "cuenta bloqueada"),
    (lambda: login("edgar", "1234", [("a", "b")]), "acceso denegado"),
    (lambda: login("edgar", "1234", []), "acceso denegado"),
])
````

<details>
<summary>🔑 Solución del ejercicio 54 — <em>ábrela solo después de intentarlo</em></summary>

```python
MAX_INTENTOS = 3

def login(usuario, contrasena, intentos):
    """Valida credenciales con un tope de intentos."""
    usados = 0
    for u, p in intentos[:MAX_INTENTOS]:      # la rebanada aplica el tope
        usados += 1
        if u == usuario and p == contrasena:
            return "acceso concedido"
    if usados >= MAX_INTENTOS:
        return "cuenta bloqueada"
    return "acceso denegado"

print(login("edgar", "1234", [("x","y"), ("edgar","1234")]))

# La versión real, con input(). No se puede probar automáticamente,
# y por eso en este curso el patrón es: lógica testeable + entrada aparte.
def login_interactivo(usuario, contrasena):
    for intento in range(MAX_INTENTOS):
        u = input("Usuario: ")
        p = input("Contraseña: ")
        if u == usuario and p == contrasena:
            return "acceso concedido"
        print(f"Incorrecto. Te quedan {MAX_INTENTOS - intento - 1} intentos.")
    return "cuenta bloqueada"
```

**Por qué:** Dos ideas de diseño aquí. Primera: `intentos[:MAX_INTENTOS]` implementa el tope sin un solo contador extra — rebanar es más seguro que llevar la cuenta a mano. Segunda, y más importante: **separa la lógica de la entrada de datos**. `login()` es pura y se prueba con seis casos en un segundo; `login_interactivo()` solo pide datos y delega. Esa separación es la razón por la que en este notebook casi no aparece `input()`, y es la misma que hace testeable cualquier programa serio.

</details>

Junta dos funciones que ya escribiste en este módulo: `solo_positivos` (ejercicio 48) y `maximo` (ejercicio 51).

## Módulo 6 — Listas y tuplas

**Ejercicios 56–68** · duración aproximada: 70–110 min

**Al terminar este módulo podrás:**

- Crear, indexar, rebanar y modificar listas

- Distinguir cuándo usar una tupla (inmutable) en lugar de una lista
- Distinguir `is` de `==` y detectar alias de objetos

### Teoría

Una **lista** (*list*) es una secuencia ordenada y **mutable**. Es la estructura de datos que más vas a usar.

```python
paises = ["MX", "BR", "CL"]
paises[0]        # "MX"      indexado
paises[-1]       # "CL"      desde el final
paises[0:2]      # ["MX","BR"]  rebanada -> lista NUEVA
paises[1] = "AR" # mutación en el lugar
```

Los métodos, con su trampa correspondiente:

| Método | Qué hace | Devuelve |
|---|---|---|
| `.append(x)` | agrega al final | `None` |
| `.insert(i, x)` | inserta en la posición `i` | `None` |
| `.remove(x)` | borra la primera aparición de `x` | `None` (y `ValueError` si no está) |
| `.pop()` / `.pop(i)` | saca el último / el `i`-ésimo | **el elemento** |
| `.sort()` | ordena en el lugar | `None` |
| `sorted(lista)` | *función*, no método | una lista **nueva** ordenada |

La regla que explica toda esa columna: en Python, **lo que muta devuelve `None`** y lo que crea algo nuevo lo devuelve. Por eso `lista = lista.append(x)` destruye tu lista y `lista = lista.sort()` la convierte en `None`.

El concepto más resbaloso del módulo es **referencia vs. copia** (*aliasing*). `b = a` **no copia**: crea un segundo nombre para el mismo objeto, y mutar `b` cambia `a`. Para copiar: `a.copy()`, `a[:]` o `list(a)` — y eso es una copia *superficial* (*shallow*): si la lista contiene listas, esas siguen compartidas.

Una **tupla** (*tuple*) es una secuencia **inmutable**: `(pais, anio, valor)`. Se usa cuando el conjunto de valores es un registro conceptualmente fijo. Al ser inmutable, puede ser llave de un diccionario; una lista no.

````python
# ---- Mini-lección: listas y tuplas ----

gastos = [120, 340, 95, 780, 210]
print(len(gastos), sum(gastos), min(gastos), max(gastos))
print(sum(gastos) / len(gastos))

# lo que muta devuelve None
print("sorted() ->", sorted(gastos), "| original:", gastos)
print(".sort()  ->", gastos.sort(), "| original:", gastos)   # ¡None y lista mutada!

# referencia vs copia: el bug clásico
a = [1, 2, 3]
b = a                 # NO es copia
c = a.copy()          # sí lo es
b.append(99)
print("a:", a, "| b:", b, "| c:", c, "| a is b:", a is b)

# tuplas: inmutables a propósito
registro = ("Guanajuato", 2024, 4.7)
estado, anio, inflacion = registro          # desempaquetado
print(f"{estado} ({anio}): {inflacion}%")

try:
    registro[2] = 5.0                        # error DELIBERADO, para verlo explicado
except TypeError as e:
    print("TypeError esperado ->", e)

# lista de tuplas: la tabla más simple posible
tabla = [("MX", 4.7), ("BR", 4.6), ("AR", 211.4)]
for pais, inf in tabla:
    print(f"{pais:>3} {inf:>8.1f}%")
````

> 🔮 **Predice antes de correr.** Después de `b = a` y `b.append(99)`, ¿cuántos elementos tiene `a`? Es la predicción más importante de todo el notebook.

```python
    """Devuelve copias ordenadas SIN tocar la lista original."""

    descendente = sorted(lista, reverse=True)
```

**Por qué:** La convención de Python es transparente una vez que la ves: los métodos que **mutan** devuelven `None` (`sort`, `append`, `reverse`), y las funciones que **crean algo nuevo** lo devuelven (`sorted`, `reversed`). `mi_lista = mi_lista.sort()` es el bug que produce esa regla.

```python
def media(datos):
    """Media aritmética."""
    if not datos:
        return None
    return sum(datos) / len(datos)

def mediana(datos):
    """Mediana: valor central de los datos ORDENADOS."""
    if not datos:
        return None
    orden = sorted(datos)          # copia ordenada; no mutamos la entrada
    n = len(orden)
    medio = n // 2
    if n % 2 == 1:
        return orden[medio]                          # n impar: hay un centro exacto
    return (orden[medio - 1] + orden[medio]) / 2     # n par: promedio de los dos centrales

print(media([1, 2, 3, 1000]), mediana([1, 2, 3, 1000]))   # 251.5  2.5
```

**Por qué:** El último `print` es la lección estadística: la media es sensible a valores extremos y la mediana no (es *robusta*). Por eso el ingreso de un país se reporta con mediana y el PIB con suma — usar media de ingreso oculta la desigualdad.

```python
def moda(datos):
    """Valor más frecuente; en caso de empate, el primero en aparecer."""
    if not datos:
        return None
    mejor = datos[0]
    mejor_conteo = datos.count(mejor)
    for valor in datos:
        c = datos.count(valor)
        if c > mejor_conteo:      # > estricto conserva al primero en empates
            mejor, mejor_conteo = valor, c
    return mejor
```

**Por qué:** Esta implementación es $O(n^2)$ porque `.count()` recorre la lista completa en cada vuelta. Con 100 datos da igual; con 100,000 se vuelve inaceptable. En el ejercicio 72 la reescribirás con un diccionario en $O(n)$ — es el mismo problema resuelto con la estructura correcta.

```python
def transponer(matriz):
    """Transpuesta de una matriz representada como lista de listas."""
    if not matriz:
        return []
    n_filas = len(matriz)
    n_cols = len(matriz[0])
    resultado = []
    for j in range(n_cols):           # cada columna original será una fila nueva
        fila_nueva = []
        for i in range(n_filas):
            fila_nueva.append(matriz[i][j])
        resultado.append(fila_nueva)
    return resultado

# Truco idiomático (Módulo 10): list(zip(*matriz)) hace lo mismo con tuplas
print([list(t) for t in zip(*[[1,2,3],[4,5,6]])])
```

**Por qué:** `matriz[i][j]` es "del elemento `i` de la lista externa, toma el elemento `j`". Esta representación funciona pero es lenta para álgebra lineal seria: en el Módulo 12 lo harás con `array.T` de NumPy, que es una operación de costo casi cero.

</details>

```python
b = a                     # dos nombres, UN objeto
b.append(3)
print(a, a is b)          # [1, 2, 3] True  -> "a" cambió sin que lo tocaras
```

**Por qué:** `b = a` copia la **referencia**, no el contenido: `a is b` es `True`. Una copia con `.copy()` es *superficial* (*shallow*): copia el primer nivel, pero si la lista contiene listas, esas siguen compartidas. Para independencia total se usa `copy.deepcopy()`.

</details>

<details>
<summary>🔑 Solución del ejercicio 67 — <em>ábrela solo después de intentarlo</em></summary>

```python
def duplicar_mal(fila, n):
    return [fila] * n

def duplicar(fila, n):
    """n filas con el mismo CONTENIDO pero objetos independientes."""
    return [fila.copy() for _ in range(n)]     # una copia nueva por renglón

m = duplicar([0, 0], 3)
m[0][0] = 9
print(m)                      # [[9, 0], [0, 0], [0, 0]]

# == compara contenido; is compara identidad
f = duplicar([0], 2)
print(f[0] == f[1], f[0] is f[1])          # True False  -> iguales pero distintos objetos
mal = duplicar_mal([0], 2)
print(mal[0] == mal[1], mal[0] is mal[1])  # True True   -> ¡es el MISMO objeto!

# id() muestra la dirección del objeto: la prueba definitiva
print(id(f[0]) == id(f[1]), id(mal[0]) == id(mal[1]))
```

**Por qué:** `[fila] * n` construye una lista de `n` **punteros al mismo objeto**, así que mutar uno muta todos. Con inmutables (`[0] * 3`) el problema no existe, porque no puedes mutar un `int`; solo aparece con listas, diccionarios y objetos. La regla operativa: **usa `is` únicamente contra `None`** (`if x is None:`) y para diagnosticar alias como aquí; para comparar valores, siempre `==`. Y cuidado: `duplicar` hace copia *superficial*, así que si `fila` contuviera listas necesitarías `copy.deepcopy`.

</details>

### Teoría

Un **diccionario** (*dict*) asocia **llaves** con **valores**. Internamente es una tabla *hash*, así que buscar por llave cuesta $O(1)$ en promedio, sin importar el tamaño.

```python
pibs = {"MX": 1790.0, "BR": 2170.0}
pibs["CL"] = 340.0          # crear o actualizar: misma sintaxis
pibs["MX"]                  # 1790.0  (KeyError si no existe)
pibs.get("AR", 0.0)         # 0.0     (nunca lanza error)
"BR" in pibs                # True
```

Recorrido: `for k in d` da llaves; `for v in d.values()` da valores; `for k, v in d.items()` da pares. Desde Python 3.7 el orden de recorrido es el de inserción, **garantizado por el lenguaje**.

Dos patrones que valen su peso en oro:

```python
# tabla de frecuencias, O(n)
conteo[x] = conteo.get(x, 0) + 1

# agrupación (el groupby hecho a mano)
grupos.setdefault(llave, []).append(x)
```

Las llaves deben ser **inmutables** (texto, número, tupla) porque su *hash* no puede cambiar. Una lista no puede ser llave; una tupla sí.

Un **conjunto** (*set*) es una colección **sin orden y sin duplicados**: `{"MX", "BR"}`. Sirve para dos cosas: eliminar duplicados (`set(lista)`) y hacer álgebra de conjuntos con `&` (intersección), `|` (unión) y `-` (diferencia). Esas tres operaciones responden preguntas de conciliación de datos que aparecen todos los días: qué claves están en ambos sistemas, cuál es el universo completo, qué falta en el otro registro.

Cuidado: `{}` es un diccionario vacío, no un conjunto. El conjunto vacío es `set()`.

````python
# ---- Mini-lección: diccionarios y conjuntos ----

pibs = {"MX": 1790.0, "BR": 2170.0, "CL": 340.0}

for pais, pib in pibs.items():
    print(f"{pais}: {pib:>8,.0f}")

print(pibs.get("AR", "no disponible"))
print("AR" in pibs, len(pibs), sum(pibs.values()))

# frecuencias en O(n)
categorias = ["comida", "transporte", "comida", "utiles", "comida"]
conteo = {}
for c in categorias:
    conteo[c] = conteo.get(c, 0) + 1
print(conteo)

# agrupación
grupos = {}
for pais, pib in pibs.items():
    tamano = "grande" if pib > 1000 else "chico"
    grupos.setdefault(tamano, []).append(pais)
print(grupos)

# diccionario anidado (así llega el JSON de una API)
inflacion = {"MX": {2023: 5.5, 2024: 4.7}, "BR": {2023: 4.6}}
print(inflacion.get("MX", {}).get(2024))
print(inflacion.get("AR", {}).get(2024))     # None, sin reventar

# conjuntos
sistema_a = {"1101", "1102", "1103"}
sistema_b = {"1102", "1103", "1104"}
print("en ambos:  ", sorted(sistema_a & sistema_b))
print("universo:  ", sorted(sistema_a | sistema_b))
print("solo en A: ", sorted(sistema_a - sistema_b))
````

```python
def total_y_promedio(pibs):
    """Suma y promedio de los valores de un diccionario."""
    total = 0
    for pais, valor in pibs.items():   # items() da pares (llave, valor)
        total += valor
    if not pibs:
        return 0, None
    return total, total / len(pibs)

print(total_y_promedio({"MX": 1.7, "BR": 2.2, "AR": 0.6}))
```

```python
def invertir(d):
    """Invierte llaves y valores. Si hay valores repetidos, gana el último."""
    resultado = {}
    for llave, valor in d.items():
        resultado[valor] = llave
    return resultado

print(invertir({"MX": 52, "BR": 55}))
print(invertir({"a": 1, "b": 1}))   # {1: "b"} -- ¡se perdió "a"!
```

```python
def agrupar_por_longitud(palabras):
    """Agrupa palabras por su número de caracteres."""
    grupos = {}
    for p in palabras:
        grupos.setdefault(len(p), []).append(p)
    return grupos

# Versión explícita, si setdefault te resulta opaco:
def agrupar_v2(palabras):
    grupos = {}
    for p in palabras:
        if len(p) not in grupos:
            grupos[len(p)] = []
        grupos[len(p)].append(p)
    return grupos
```

**Por qué:** Este es el `groupby` de SQL y de pandas, hecho a mano. `setdefault` devuelve la lista (nueva o existente) y sobre ella aplicas `.append()` directamente. La alternativa moderna es `collections.defaultdict(list)`.

```python
    """Cuenta y ordena los valores distintos."""
    distintos = set(lista)             # los duplicados desaparecen
```

```python
def comparar(a, b):
    """Álgebra de conjuntos entre dos listas."""
    sa, sb = set(a), set(b)
    interseccion = sorted(sa & sb)   # también sa.intersection(sb)
    union = sorted(sa | sb)          # también sa.union(sb)
    solo_a = sorted(sa - sb)         # también sa.difference(sb)
    return interseccion, union, solo_a

print(comparar([1,2,3], [2,3,4]))
```

**Por qué:** Estas tres operaciones responden preguntas de conciliación de datos que aparecen a diario: ¿qué claves presupuestales están en ambos sistemas? (intersección), ¿cuál es el universo completo? (unión), ¿qué se registró en el mío y no en el otro? (diferencia). Hacerlo con ciclos anidados sería $O(n \cdot m)$; con conjuntos es lineal.

```python
def reporte_encuesta(respuestas):
    """Reporte agregado de una encuesta de gasto."""
    if not respuestas:
        return None

    categorias = [r["categoria"] for r in respuestas]
    gastos = [r["gasto"] for r in respuestas]

    conteo = frecuencias(categorias)                    # función del módulo de diccionarios
    mas_comun = max(conteo, key=conteo.get)             # en empate gana la primera insertada

    por_categoria = {}
    for r in respuestas:
        c = r["categoria"]
        por_categoria[c] = por_categoria.get(c, 0) + r["gasto"]

    return {
        "n": len(respuestas),
        "categoria_mas_comun": mas_comun,
        "gasto_total": sum(gastos),
        "gasto_promedio": round(sum(gastos) / len(gastos), 2),
        "por_categoria": por_categoria,
    }

import json
print(json.dumps(reporte_encuesta([
    {"categoria": "comida", "gasto": 120},
    {"categoria": "transporte", "gasto": 45},
    {"categoria": "comida", "gasto": 80},
]), indent=2, ensure_ascii=False))
```

## Módulo 8 — Funciones II — herramientas de verdad

**Ejercicios 80–89** · duración aproximada: 70–110 min

**Al terminar este módulo podrás:**

- Evitar estado global y escribir funciones idempotentes
- Documentar con docstrings y anotar tipos
- Escribir funciones recursivas con caso base y entender su costo
- Usar `*args`, `**kwargs`, funciones de orden superior y clausuras
- Reconocer el bug del argumento por defecto mutable e implementar memoización

### Teoría

Ahora que ya usaste funciones para todo, vale la pena aprender a escribirlas bien.

**Ámbito** (*scope*). Cada llamada crea un espacio de nombres local. Python resuelve nombres con la regla **LEGB**: Local → Enclosing → Global → Built-in. Una función puede *leer* una variable global, pero para *escribirla* necesita la palabra `global` — y si te encuentras escribiéndola, casi siempre lo que querías era un `return`. El criterio: una función debe ser **idempotente**, o sea que llamarla dos veces con la misma entrada dé el mismo resultado.

**Docstrings y anotaciones de tipo.** El texto entre triples comillas justo debajo de la firma queda accesible en `f.__doc__` y en `help(f)`. Las anotaciones (`def f(x: float) -> float:`) no se verifican en ejecución: son documentación que tu editor y `mypy` sí leen.

**Recursión.** Una función que se llama a sí misma necesita (1) un **caso base** que no recurre y (2) que cada llamada se acerque a él. Python limita la profundidad a ~1000 marcos. La recursión ingenua de Fibonacci es $O(\varphi^n)$ porque recalcula subproblemas; `@lru_cache` la vuelve $O(n)$.

**Argumentos flexibles.** `*args` empaqueta los posicionales sobrantes en una tupla; `**kwargs`, los nombrados en un diccionario. En la *llamada*, el mismo asterisco hace lo inverso: desempaqueta (`f(*lista)`).

**Orden superior y clausuras.** Las funciones son objetos de primera clase: se guardan, se pasan y se devuelven. Una **clausura** (*closure*) es una función interna que recuerda las variables de la externa incluso después de que esta terminó — la forma limpia de fabricar familias de funciones parametrizadas.

````python
# ---- Mini-lección: funciones II ----
from functools import lru_cache
import time

# Ámbito: la función NO ve el interior de otra, y no debe escribir globales
tasa_global = 0.16

def con_iva(precio):
    """Lee la global (aceptable) pero no la modifica (importante)."""
    return precio * (1 + tasa_global)

print(con_iva(100))

# Recursión con y sin memoización: mide la diferencia
def fib_lento(n):
    return n if n < 2 else fib_lento(n-1) + fib_lento(n-2)

@lru_cache(maxsize=None)
def fib_rapido(n):
    return n if n < 2 else fib_rapido(n-1) + fib_rapido(n-2)

t0 = time.perf_counter(); fib_lento(28);  t1 = time.perf_counter()
t2 = time.perf_counter(); fib_rapido(28); t3 = time.perf_counter()
print(f"ingenua: {t1-t0:.4f}s | memoizada: {t3-t2:.8f}s")

# *args y **kwargs
def resumen(*valores, **opciones):
    decimales = opciones.get("decimales", 2)
    return f"n={len(valores)} media={sum(valores)/len(valores):.{decimales}f}"

print(resumen(1, 2, 3))
print(resumen(1, 2, 3, decimales=4))

# Orden superior + clausura
def crear_ajustador(factor):
    def ajustar(x):
        return x * factor
    return ajustar

inflar = crear_ajustador(1.047)
print([round(inflar(v), 2) for v in (100, 200, 300)])
````

```python
    total = 0            # variable local: nace y muere con cada llamada
```

```python
def varianza_muestral(datos: list) -> float:
    """Varianza muestral (con corrección de Bessel, n-1).

    Parámetros
    ----------
    datos : list
        Lista de números. Requiere al menos 2 observaciones.

    Devuelve
    --------
    float o None si hay menos de 2 datos.
    """
    n = len(datos)
    if n < 2:
        return None
    media = sum(datos) / n
    suma_cuadrados = 0.0
    for x in datos:
        suma_cuadrados += (x - media) ** 2
    return suma_cuadrados / (n - 1)

print(varianza_muestral([2,4,4,4,5,5,7,9]))
help(varianza_muestral)
```

**Por qué:** Las anotaciones de tipo (*type hints*) **no** se verifican en tiempo de ejecución: son documentación que herramientas como `mypy` y tu editor sí revisan. El $n-1$ es la corrección de Bessel: hace el estimador insesgado, porque usar la media muestral ya consumió un grado de libertad.

<summary>🔑 Solución del ejercicio 82 — <em>ábrela solo después de intentarlo</em></summary>

```python
def factorial(n):
    """n! calculado recursivamente."""
    if n <= 1:              # CASO BASE: sin él, recursión infinita
        return 1
    return n * factorial(n - 1)   # el argumento se acerca al caso base

print(factorial(5))   # 5*4*3*2*1 = 120

# Versión iterativa equivalente (más eficiente en memoria):
def factorial_iterativo(n):
    r = 1
    for i in range(2, n + 1):
        r *= i
    return r
```

**Por qué:** Cada llamada recursiva ocupa un marco en la pila (*stack frame*), y Python limita la profundidad a ~1000 por defecto: `factorial(5000)` lanza `RecursionError`. La recursión se justifica cuando el problema **es** recursivo (árboles, estructuras anidadas), no para reemplazar un ciclo.

</details>

<details>
<summary>🔑 Solución del ejercicio 83 — <em>ábrela solo después de intentarlo</em></summary>

```python
def fibonacci(n):
    """n-ésimo número de Fibonacci (versión recursiva ingenua)."""
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

# Versión con memoización: de exponencial a lineal con una línea
from functools import lru_cache

@lru_cache(maxsize=None)
def fib_rapido(n):
    return n if n < 2 else fib_rapido(n-1) + fib_rapido(n-2)

print(fibonacci(15), fib_rapido(90))
```

**Por qué:** La versión ingenua es $O(\varphi^n)$: `fibonacci(40)` hace más de 300 millones de llamadas porque recalcula los mismos subproblemas una y otra vez. `@lru_cache` guarda cada resultado (*memoización*) y lo baja a $O(n)$. Es el ejemplo canónico de por qué la complejidad algorítmica importa más que la velocidad del lenguaje.

</details>

```python
def promedio(*valores):
    """Media de un número arbitrario de argumentos."""
    if not valores:          # valores es una tupla, vacía si no pasaron nada
        return 0.0
    return sum(valores) / len(valores)

print(promedio(1, 2, 3))       # 2.0
print(promedio(*[2, 4, 6, 8])) # 5.0 -- el * también DESEMPAQUETA una lista al llamar
```

**Por qué:** El asterisco tiene dos caras: en la **definición** empaqueta (`*args` recoge lo suelto en una tupla) y en la **llamada** desempaqueta (`f(*lista)` reparte los elementos como argumentos separados). Es el mismo símbolo haciendo operaciones inversas según dónde aparezca.

```python
def aplicar(funcion, lista):
    """Aplica una función a cada elemento (equivalente a map)."""
    resultado = []
    for x in lista:
        resultado.append(funcion(x))   # sin paréntesis sería la función misma, no su resultado

print(aplicar(abs, [-1, 2, -3]))

```

<details>
<summary>🔑 Solución del ejercicio 87 — <em>ábrela solo después de intentarlo</em></summary>

```python
def crear_deflactor(indice_base):
    """Fabrica una función que deflacta valores nominales con un índice fijo."""
    def deflactar(valor_nominal):
        # indice_base sigue vivo aquí aunque crear_deflactor ya terminó: eso es una clausura
        return valor_nominal / indice_base * 100
    return deflactar        # SIN paréntesis: devolvemos la función, no su resultado

deflactar_2024 = crear_deflactor(110)
print(deflactar_2024(220))       # 200.0
print([deflactar_2024(v) for v in (110, 220, 330)])
```

**Por qué:** La función interna "recuerda" `indice_base` incluso después de que la externa terminó: ese entorno capturado es la **clausura**. Es la forma más limpia de crear familias de funciones parametrizadas (un deflactor por año base) sin repetir código ni usar variables globales.

</details>

```python
    historial.append(gasto)
    return historial

def agregar_gasto(gasto, historial=None):
    """Agrega un gasto. Si no hay historial, crea uno NUEVO en cada llamada."""
    if historial is None:      # centinela: None marca "no me pasaron nada"
        historial = []
    historial.append(gasto)
    return historial
```

```python
def con_cache(funcion):
    """Envuelve una función de un argumento con memoización manual."""
    cache = {}

    def envuelta(x):
        if x not in cache:              # solo calcula si no lo tiene guardado
            cache[x] = funcion(x)
        return cache[x]

    envuelta.cache = cache              # las funciones aceptan atributos
    return envuelta

import time

def fib_lento(n):
    return n if n < 2 else fib_lento(n - 1) + fib_lento(n - 2)

fib_c = con_cache(fib_lento)
t0 = time.perf_counter(); fib_c(27); t1 = time.perf_counter()
t2 = time.perf_counter(); fib_c(27); t3 = time.perf_counter()   # ya está en cache
print(f"1ª llamada: {t1-t0:.4f}s | 2ª llamada: {t3-t2:.8f}s | cache: {list(fib_c.cache)}")
```

**Por qué:** Acabas de escribir un **decorador** sin saberlo: `@lru_cache` es esto mismo, más un tope de tamaño, soporte para varios argumentos y una implementación en C. Dos límites reales de tu versión que vale la pena ver: solo acepta un argumento, y ese argumento debe ser *hashable* (una lista como entrada lanzaría `TypeError: unhashable type`). Y una advertencia: memoizar una función que depende de algo externo (la hora, un archivo, la red) devuelve resultados obsoletos — la memoización solo es segura sobre funciones **puras**.

</details>

## Módulo 9 — Errores y robustez

**Ejercicios 90–99** · duración aproximada: 60–95 min

**Al terminar este módulo podrás:**

- Leer un traceback e identificar el tipo de excepción
- Capturar excepciones específicas con `try` / `except` / `else` / `finally`
- Lanzar errores propios con `raise` cuando la entrada es inválida
- Distinguir `assert` (invariante de programador) de validación de datos
- Armar un pipeline de limpieza que no invente datos

### Teoría

Programar es, en buena medida, leer errores. Un **traceback** se lee **de abajo hacia arriba**: la última línea trae el tipo y el mensaje, y las de arriba la cadena de llamadas que llegó ahí.

Excepciones que verás todos los días:

| Excepción | Significado típico |
|---|---|
| `SyntaxError` | el código no es Python válido (se detecta antes de correr) |
| `NameError` | nombre no definido: un typo, o una celda que no corriste |
| `TypeError` | operación entre tipos incompatibles (`"3" + 4`) |
| `ValueError` | tipo correcto, valor imposible (`int("abc")`) |
| `IndexError` / `KeyError` | posición o llave que no existe |
| `ZeroDivisionError` | división entre cero |
| `AttributeError` | el objeto no tiene ese método (`(1,2).append(3)`) |

La estructura completa:

```python
try:
    ...            # código que puede fallar
except ValueError as e:
    ...            # plan B para ESE error; e trae el mensaje
except (TypeError, KeyError):
    ...            # varios tipos en un solo except
else:
    ...            # solo si NO hubo excepción
finally:
    ...            # SIEMPRE, con o sin error (cerrar archivos, conexiones)
```

Python prefiere el estilo **EAFP** (*Easier to Ask Forgiveness than Permission*): intenta y captura, en vez de verificar todo antes. Lo que nunca debes escribir es `except:` desnudo o `except Exception: pass`: atrapa hasta un `Ctrl+C` y esconde bugs reales.

`raise ValueError("mensaje")` genera un error a propósito, y es lo correcto cuando la entrada es **inválida**: falla exactamente donde está el problema en lugar de devolver `None` y explotar 200 líneas después (*fail fast*).

`assert condicion, "mensaje"` documenta un supuesto que crees imposible de violar. Ojo: Python **desactiva** todos los `assert` con la bandera `-O`, así que nunca los uses para validar datos de producción.

En este notebook tienes disponible la función auxiliar `tipo_de_error(f, *args)`, que devuelve el nombre de la excepción que lanza `f` — la usan varias verificaciones.

````python
# ---- Mini-lección: errores ----

# Cada error, provocado y explicado a propósito
pruebas = [
    ("10 / 0",              lambda: 10 / 0),
    ('"3" + 4',             lambda: "3" + 4),
    ('int("abc")',          lambda: int("abc")),
    ("[1,2,3][5]",          lambda: [1,2,3][5]),
    ('{"a":1}["b"]',        lambda: {"a":1}["b"]),
    ("(1,2).append(3)",     lambda: (1,2).append(3)),
]
for codigo, f in pruebas:
    try:
        f()
    except Exception as e:
        print(f"{codigo:20} -> {type(e).__name__}: {e}")

print()

# try / except / else / finally, los cuatro bloques
def leer_entero(texto):
    try:
        n = int(texto)
    except ValueError:
        print(f"  '{texto}' no es un entero -> uso 0")
        n = 0
    else:
        print(f"  '{texto}' convertido bien")
    finally:
        print("  (finally corre siempre)")
    return n

print(leer_entero("42"))
print(leer_entero("n/d"))

# raise: fallar temprano y con un mensaje útil
def raiz(x):
    if x < 0:
        raise ValueError(f"raíz de negativo: {x}")
    return x ** 0.5

print(tipo_de_error(raiz, -1), "|", raiz(16))
````

> 🔮 **Predice antes de correr.** ¿`int("3.5")` lanza `ValueError` o devuelve `3`? Y `float("3.5")`, ¿qué hace?

<summary>🔑 Solución del ejercicio 91 — <em>ábrela solo después de intentarlo</em></summary>

```python
def dividir(a, b):
    """División protegida: devuelve None si el divisor es cero."""
    try:
        return a / b
    except ZeroDivisionError:
        return None

print(dividir(10, 2), dividir(10, 0))
```

**Por qué:** Existen dos filosofías: *LBYL* ("mira antes de saltar", con `if b != 0`) y *EAFP* ("pide perdón, no permiso", con `try/except`). Python prefiere la segunda, porque evita duplicar la verificación y no tiene condiciones de carrera. Lo que **nunca** debes hacer es `except:` desnudo: atrapa hasta un `Ctrl+C`.

</details>

<details>
<summary>🔑 Solución del ejercicio 92 — <em>ábrela solo después de intentarlo</em></summary>

```python
def convertir(texto):
    """Convierte texto a entero reportando el estado de la operación."""
    try:
        numero = int(texto)
    except ValueError:
        return "error", 0
    else:
        # corre SOLO si el try no lanzó excepción
        return "ok", numero
    finally:
        # corre SIEMPRE, con o sin error: ideal para cerrar archivos o conexiones
        pass

print(convertir("42"), convertir("3.5"))   # ("ok", 42) ("error", 0)
```

**Por qué:** `int("3.5")` falla porque `"3.5"` no es un entero literal (habría que hacer `int(float("3.5"))`). El valor de `else` es mantener el `try` reducido a la línea que de verdad puede fallar; `finally` se ejecuta pase lo que pase, incluso si hay un `return` dentro del `try`.

</details>

```python
def obtener(coleccion, llave):
    """Acceso genérico que reporta qué tipo de error ocurrió."""
    try:
        return coleccion[llave]
    except KeyError:
        return "KeyError"      # diccionario sin esa llave
    except IndexError:
        return "IndexError"    # lista con índice fuera de rango
    except TypeError:
        return "TypeError"     # índice de tipo inválido (lista["a"])

print(obtener({"a": 1}, "b"), obtener([1,2], 5), obtener([1,2], "a"))
```

**Por qué:** Los bloques `except` se evalúan en orden y solo uno se ejecuta. Como `KeyError` e `IndexError` son ambos subclases de `LookupError`, si pusieras `except LookupError` primero atraparía los dos — por eso el orden va de lo específico a lo general.

</details>

<summary>🔑 Solución del ejercicio 97 — <em>ábrela solo después de intentarlo</em></summary>

```python
def promedio_mal(datos):
    try:
        return sum(datos) / len(datos)
    except:
        return 0

def promedio_seguro(datos):
    """Promedio. None si la lista está vacía; el TypeError se propaga a propósito."""
    if not datos:
        return None
    return sum(datos) / len(datos)     # si hay texto, TypeError sube y AVISA

print("mal   :", promedio_mal(["a", "b"]), "<- un cero inventado, sin avisar")
print(promedio_seguro([1, 2, 3]), promedio_seguro([]))
print(tipo_de_error(promedio_seguro, ["a", "b"]))   # TypeError: correcto, no lo escondemos

# Por qué el except desnudo es peor de lo que parece:
#   except:            atrapa TODO, incluido KeyboardInterrupt y SystemExit
#   except Exception:  atrapa casi todo, pero al menos deja pasar Ctrl+C
#   except ValueError: atrapa solo lo que sabes manejar  <- esto es lo correcto
```

**Por qué:** Un `except:` desnudo atrapa `BaseException`, así que también intercepta el `Ctrl+C` del usuario y el `SystemExit`: puedes terminar con un programa que **no se puede detener**. Pero el daño mayor es el silencioso: convierte un bug ruidoso (`TypeError`, que te dice exactamente qué pasó) en un dato falso (`0`) que se propaga a un promedio, a una gráfica y a una conclusión. La regla es capturar **solo** la excepción que sabes manejar, y dejar subir todo lo demás.

</details>

#### Ejercicio 98 — Escribe tú las pruebas

**Concepto:** Diseño de casos de prueba — *test design*

Llevas 97 ejercicios corriendo verificaciones que escribí yo. Este las invierte: **ahora las escribes tú**, y el notebook comprueba que estén bien pensadas.

Tu `promedio_seguro` del ejercicio 97 devuelve el promedio de una lista, `None` si está vacía, y deja pasar el `TypeError` si los elementos no son números.

Construye `mis_casos`: una **lista de parejas** `(entrada, esperado)`, donde `entrada` es la lista que le pasarías y `esperado` es lo que **debe** devolver. Requisitos:

No incluyas el caso de los elementos no numéricos: ese lanza una excepción en lugar de devolver algo, y probar excepciones necesita otra forma (la que usa `tipo_de_error`).

Piensa en familias, no en ejemplos sueltos: el caso **normal** (varios números), el caso **frontera** (un solo elemento, la lista vacía) y el caso **incómodo** (negativos, decimales, un promedio que da exactamente cero). Escribe el `esperado` **a mano**, sin correr la función: de eso se trata.

</details>

````python
# ---------- Ejercicio 98: TU SOLUCIÓN ----------
# Cada pareja es (entrada, esperado). Calcula el esperado TÚ, no lo copies de la salida.
mis_casos = [
    ([...], ...),
    ([...], ...),
    ([...], ...),
    ([...], ...),
]
````

````python
# ---------- Verificación (no la edites) ----------
def _casos_correctos(casos):
    """True si cada 'esperado' coincide con lo que devuelve promedio_seguro."""
    for entrada, esperado in casos:
        real = promedio_seguro(entrada)
        if real is None or esperado is None:
            if not (real is None and esperado is None):
                return False
        elif abs(real - esperado) > 1e-9:
            return False
    return True

probar("Ejercicio 98", [
    (lambda: len(mis_casos) >= 4, True),
    (lambda: _casos_correctos(mis_casos), True),
    (lambda: any(list(e) == [] for e, _ in mis_casos), True),
    (lambda: any(len(e) == 1 for e, _ in mis_casos), True),
    (lambda: len({tuple(e) for e, _ in mis_casos}) == len(mis_casos), True),
])
````

<details>
<summary>🔑 Solución del ejercicio 98 — <em>ábrela solo después de intentarlo</em></summary>

```python
mis_casos = [
    ([2, 4, 6],    4.0),    # normal: varios números
    ([10],        10.0),    # frontera: un solo elemento
    ([],          None),    # frontera: lista vacía
    ([-1, 1],      0.0),    # incómodo: promedio exactamente cero
    ([0.5, 1.5],   1.0),    # incómodo: decimales
]

    print(entrada, "-> esperaba", esperado, "y dio", promedio_seguro(entrada))
```

**Por qué:** Un buen conjunto de casos no se mide por cuántos son, sino por **cuántas formas distintas de fallar cubre**. Los cinco de arriba cubren cinco:

| Caso | Qué bug atraparía |
|---|---|
| `[2, 4, 6]` | que la función simplemente no calcule bien |
| `[10]` | dividir entre `len - 1`, o suponer que siempre hay al menos dos |
| `[]` | `ZeroDivisionError` por no revisar la lista vacía |
| `[-1, 1]` | confundir "promedio cero" con "sin datos" — el clásico `if not resultado:` |
| `[0.5, 1.5]` | forzar `int` en algún punto del cálculo |

Ese cuarto caso es el más instructivo. Un promedio de `0.0` es un resultado **válido**, pero `0.0` es falso en un contexto booleano, igual que `None` y que `[]`. Una función que devuelve `None` para "sin datos" y `0.0` para "promedio cero" obliga a quien la llama a distinguirlas con `is None`, no con `if resultado:`. Si nunca pruebas un promedio de cero, ese bug vive en tu código durante meses.

Y fíjate en el requisito de las entradas distintas: cuatro casos iguales son un solo caso escrito cuatro veces. Es la trampa más común al escribir pruebas para cumplir una cuota en lugar de para encontrar errores.

```python
def limpiar_y_resumir(crudos):
    """Convierte, descarta lo inconvertible y resume lo que quedó."""
    convertidos = [a_float(v, None) for v in crudos]           # None = no se pudo
    validos = [v for v in convertidos if v is not None]        # OJO: "is not None", no "if v"

    if not validos:
        return None

    return {
        "n_validos": len(validos),
        "n_descartados": len(crudos) - len(validos),
        "media": round(media(validos), 2),
        "mediana": mediana(validos),
    }

sucios = ["12", "n/d", 15, None, "9.5", "", "1,200"]
print(limpiar_y_resumir(sucios))

# Nota "1,200": float("1,200") falla porque la coma no es separador de miles válido.
# En datos mexicanos esto es MUY común. Se arregla antes de convertir:
print(a_float("1,200".replace(",", ""), None))   # 1200.0
```

**Por qué:** El detalle del `default=None` en lugar de `0.0` es el punto entero del ejercicio: con `0.0`, los siete valores sucios se vuelven siete ceros, `n_validos` miente y la media se hunde, **sin un solo error en pantalla**. Y el filtro tiene su propia trampa: `if v` descarta el `0.0` legítimo porque cero es falso en Python; hay que escribir `if v is not None`. Estas dos decisiones —qué centinela usar y cómo filtrarlo— son la diferencia entre un pipeline correcto y uno que produce números plausibles y falsos.

</details>

## Módulo 10 — Estilo pythonic y librería estándar

**Ejercicios 100–108** · duración aproximada: 60–95 min

**Al terminar este módulo podrás:**

- Escribir comprensiones de listas y diccionarios
- Ordenar con `key=` y usar `lambda` con criterio
- Conocer `map`/`filter` y saber por qué normalmente no los usarás
- Usar `math`, `random` (con semilla) y `collections.Counter`
- Entender por qué `0.1 + 0.2 != 0.3` y cuándo un generador se agota

### Teoría

Ya sabes programar. Este módulo es sobre escribir el mismo programa de forma más corta, más rápida y más legible — lo que la comunidad llama código **pythonic**.

Una **comprensión de lista** (*list comprehension*) es la versión declarativa del patrón "lista vacía + ciclo + append":

```python
[expresion for x in iterable if condicion]
{llave: valor for k, v in d.items()}      # comprensión de diccionario
(x for x in datos)                        # expresión generadora: perezosa, sin memoria
```

Es más rápida que el ciclo equivalente porque la iteración corre en C. La regla de estilo: si necesitas más de dos `for` o un `if` complicado, vuelve al ciclo explícito.

**`lambda`** es una función anónima de una sola expresión. Su uso legítimo es como argumento `key`:

```python
sorted(pares, key=lambda t: t[1], reverse=True)
```

`map` y `filter` existen y son herencia funcional, pero en Python la comprensión gana en legibilidad casi siempre. Vale conocerlos porque los verás en código ajeno.

De la **librería estándar** (*standard library*), tres módulos con los que ya puedes hacer trabajo real:

- `math`: `sqrt`, `log`, `exp`, `pi`, `floor`, `ceil`
- `random`: `randint`, `random`, `choice`, `shuffle` — y `seed()`, que es **obligatorio** si quieres que tu simulación sea reproducible
- `collections`: `Counter` (frecuencias en una línea), `defaultdict` (agrupaciones sin `setdefault`)

La reproducibilidad no es un detalle de estilo: sin semilla fija, una simulación de Monte Carlo da un número distinto cada vez y nadie puede replicar tu resultado.

````python
# ---- Mini-lección: pythonic ----
import math, random
from collections import Counter, defaultdict

datos = [12, 7, 23, 4, 19, 8]

# Comprensiones: qué quieres, no cómo construirlo
print([x ** 2 for x in datos])
print([x for x in datos if x > 10])
print({x: x ** 0.5 for x in datos if x > 15})

# Generador: no construye la lista completa en memoria
print(sum(x ** 2 for x in datos))

# sorted con key
tabla = [("MX", 4.7), ("BR", 4.6), ("AR", 211.4), ("CL", 3.9)]
print(sorted(tabla, key=lambda t: t[1], reverse=True))
print(max(tabla, key=lambda t: t[1]))

# math
print(math.sqrt(2), math.log(100, 10), math.ceil(4.1), math.floor(4.9))

# random reproducible
random.seed(42)
a = [random.randint(1, 6) for _ in range(10)]
random.seed(42)
b = [random.randint(1, 6) for _ in range(10)]
print(a, b, a == b)     # idénticos: misma semilla

# Counter y defaultdict
categorias = ["comida", "transporte", "comida", "utiles", "comida"]
print(Counter(categorias), Counter(categorias).most_common(1))

grupos = defaultdict(list)
for pais, inf in tabla:
    grupos["alta" if inf > 10 else "baja"].append(pais)
print(dict(grupos))
````

````python
def cuadrados(n):
````

```python
def pares_al_cuadrado(lista):
    """Cuadrados de los elementos pares."""
    return [x ** 2 for x in lista if x % 2 == 0]

def aplanar(matriz):
    """Convierte una lista de listas en una lista simple."""
    return [x for fila in matriz for x in fila]   # mismo orden que el for anidado
```

```python
def top_n(pares, n):
    """Los n pares con mayor valor, de mayor a menor."""
    ordenados = sorted(pares, key=lambda t: t[1], reverse=True)
    return ordenados[:n]     # rebanar más allá del final no da error

print(top_n([("MX",4.7),("BR",9.3),("CL",3.1)], 2))

# key acepta cualquier función: por nombre, por longitud, por valor absoluto...
print(sorted([("MX",4.7),("BR",9.3)], key=lambda t: t[0]))
```

**Por qué:** `key` recibe una función que se aplica a cada elemento para decidir el criterio de orden — sin ella, `sorted` compararía tuplas completas (primero por nombre). Una `lambda` es una función anónima de una sola expresión: úsala solo para cosas así de cortas; si necesita nombre o dos líneas, usa `def`.

<summary>🔑 Solución del ejercicio 104 — <em>ábrela solo después de intentarlo</em></summary>

```python
def con_map(lista):
    """Aplica IVA con map."""
    return list(map(lambda x: x * 1.16, lista))

def con_filter(lista):
    """Filtra valores mayores a 100 con filter."""
    return list(filter(lambda x: x > 100, lista))

# Las comprensiones equivalentes, que en Python son la forma preferida:
#   [x * 1.16 for x in lista]
#   [x for x in lista if x > 100]
print(con_map([100, 200]), con_filter([50, 100, 150]))
```

**Por qué:** `map`/`filter` son herencia funcional (Lisp) y siguen siendo útiles con funciones ya nombradas (`map(str, lista)` es limpísimo). Pero con `lambda` la comprensión gana en legibilidad, y esa es la recomendación de la comunidad — Guido van Rossum llegó a proponer eliminarlas del lenguaje.

</details>

```python
import math

def desviacion_estandar(datos):
    """Desviación estándar muestral."""
    n = len(datos)
    if n < 2:
        return None
    media = sum(datos) / n
    varianza = sum((x - media) ** 2 for x in datos) / (n - 1)   # generador, sin lista intermedia
    return math.sqrt(varianza)

print(desviacion_estandar([2,4,4,4,5,5,7,9]))
print(math.pi, math.e, math.log(math.e), math.exp(1))
```

**Por qué:** Nota el argumento sin corchetes en `sum(... for x in datos)`: es una **expresión generadora**, que produce los valores uno por uno sin construir la lista completa en memoria. Con 8 datos da igual; con 80 millones es la diferencia entre correr y morir por falta de RAM.

<details>
<summary>🔑 Solución del ejercicio 106 — <em>ábrela solo después de intentarlo</em></summary>

```python
import random
from collections import Counter

def simular_dados(n, semilla):
    """Simula n lanzamientos de un dado justo, de forma reproducible."""
    random.seed(semilla)                                   # fija el estado del generador
    tiradas = [random.randint(1, 6) for _ in range(n)]     # _ = variable que no usamos
    return dict(Counter(tiradas))                           # Counter cuenta en O(n)

print(simular_dados(6000, 42))
print(simular_dados(50, 7) == simular_dados(50, 7))   # True: reproducible
```

**Por qué:** Fijar la semilla es un requisito de **reproducibilidad científica**, no un detalle: sin ella, tu simulación de Monte Carlo da un número distinto cada vez y nadie puede replicar tu tesis. Los generadores de Python son pseudoaleatorios (Mersenne Twister): deterministas dado el estado inicial. Para trabajo serio, `numpy.random.default_rng(semilla)` es la opción moderna.

</details>

````python
# ---------- Verificación (no la edites) ----------
probar("Ejercicio 107", [
    (lambda: 0.1 + 0.2 == 0.3, False),
    (lambda: casi_igual(0.1 + 0.2, 0.3), True),
    (lambda: casi_igual(1.0, 1.1), False),
    (lambda: casi_igual(0.0, 0.0), True),
    (lambda: casi_igual(1e-18, 0.0), True),
    (lambda: casi_igual(1.0, 1.05, tol=0.1), True),
])
````

<details>
<summary>🔑 Solución del ejercicio 107 — <em>ábrela solo después de intentarlo</em></summary>

```python
import math

def casi_igual(a, b, tol=1e-9):
    """Comparación con tolerancia ABSOLUTA."""
    return abs(a - b) <= tol

print(0.1 + 0.2, 0.1 + 0.2 == 0.3, casi_igual(0.1 + 0.2, 0.3))

# La herramienta correcta de la librería estándar: tolerancia RELATIVA + absoluta
print(math.isclose(0.1 + 0.2, 0.3))                      # True
print(math.isclose(1e9, 1e9 + 1))                        # True: 1 peso en mil millones
print(casi_igual(1e9, 1e9 + 1))                          # False: tolerancia absoluta

# Para DINERO la solución no es tolerar el error, es no cometerlo:
from decimal import Decimal
print(0.1 + 0.2 == 0.3)                                            # False
print(Decimal("0.1") + Decimal("0.2") == Decimal("0.3"))           # True
```

**Por qué:** Tres reglas que se derivan de esto. **Una:** en código real nunca compares `float` con `==` ni con una tolerancia puramente absoluta como la de `casi_igual`; usa `math.isclose`, cuya tolerancia relativa es la correcta cuando las magnitudes varían. Construir `casi_igual` sirvió para ver el mecanismo — es en esencia lo que hace `probar()` en este notebook. **Dos:** para dinero y cifras presupuestales usa `decimal.Decimal`, que trabaja en base 10 y no tiene el problema — es la razón por la que un sistema contable serio no usa `float`. **Tres:** el error se **acumula**: sumar 0.1 diez mil veces te aleja del resultado exacto de forma medible, así que en simulaciones largas conviene sumar en el orden que minimiza la pérdida de precisión (o usar `math.fsum`).

</details>

<summary>🔑 Solución del ejercicio 108 — <em>ábrela solo después de intentarlo</em></summary>

```python
predicciones = [
    "list",          # 1. corchetes -> lista
    "generator",     # 2. paréntesis -> generador perezoso
    6,               # 3. 0+1+2+3
    2,               # 4. los impares 1 y 3 (x % 2 es 1, que es verdadero)
    ["1", "2"],      # 5. map devuelve un iterador; list() lo materializa
    [3, 0],          # 6. el generador SE AGOTA: la segunda suma no encuentra nada
]

g = (x for x in range(3))
print(sum(g), sum(g))      # 3 0

# Un generador se consume una sola vez. Una lista, todas las que quieras:
l = [x for x in range(3)]
print(sum(l), sum(l))      # 3 3

# Por eso los generadores brillan con datos grandes, y muerden si los reusas:
import sys
print(sys.getsizeof([x for x in range(100000)]), "bytes (lista)")
print(sys.getsizeof(x for x in range(100000)), "bytes (generador)")
```

**Por qué:** El caso 6 es un bug real y silencioso: nadie lanza un error, simplemente el segundo cálculo da `0` (o una lista vacía, o un promedio de `nan`). Ocurre todo el tiempo al pasar un generador a dos funciones seguidas. La regla: si vas a recorrer los datos **más de una vez**, usa una lista; si solo una vez y son muchos, usa un generador — la diferencia de memoria en el último `print` es de cinco órdenes de magnitud.

</details>

## Módulo 11 — Objetos: qué es una clase y cuándo escribirla

**Ejercicios 109–111** · duración aproximada: 40–60 min

**Al terminar este módulo podrás:**

- Leer una clase y saber qué son `self`, un atributo y un método
- Entender por qué `texto.strip()` y `df.mean()` funcionan, en vez de tomarlo como magia
- Escribir una clase cuando hay un **invariante** que proteger
- Y, sobre todo, reconocer cuándo **no** vale la pena y basta un diccionario

### Teoría

Llevas más de cien ejercicios **usando** objetos sin que nadie te dijera qué son. Cada vez que escribiste `texto.strip()`, `lista.append(x)` o `df.mean()` estabas llamando a un método de un objeto. Este módulo cierra ese hueco, y lo hace justo antes de NumPy y pandas para que llegues a `Series` y `DataFrame` sabiendo qué estás tocando.

**Un objeto es datos y comportamiento que viajan juntos.** Una `str` no es solo una cadena de caracteres: es esa cadena **más** los cincuenta métodos que sabe hacerse a sí misma. Por eso `"  hola ".strip()` funciona y `strip("  hola ")` no existe.

**Una clase es el molde; el objeto es la pieza.**

```python
class Cuenta:
    def __init__(self, titular, saldo=0.0):   # se ejecuta al crear la pieza
        self.titular = titular                 # atributo: dato que vive en el objeto
        self.saldo = saldo

    def depositar(self, monto):                # método: función que recibe el objeto
        self.saldo += monto
        return self.saldo
```

**`self` no tiene nada de especial: es el objeto mismo.** `a.depositar(100)` es azúcar sintáctica para `Cuenta.depositar(a, 100)` — Python te pasa `a` como primer argumento. Ese es todo el misterio.

**`__init__` y `__repr__` son los dos métodos especiales que valen la pena de entrada.** El primero construye; el segundo decide qué ves cuando imprimes el objeto. Sin `__repr__` obtienes `<__main__.Cuenta object at 0x000001A2>`, que no le sirve a nadie.

**Y ahora la parte que casi ningún curso dice.** La pregunta no es *cómo* escribir una clase, sino **cuándo**. En análisis de datos, la respuesta es "menos veces de las que crees": el código idiomático son funciones que reciben y devuelven datos, como todo lo que llevas escrito. Una clase se gana su lugar cuando hay **estado que persiste entre llamadas** o un **invariante que proteger** — una regla que siempre debe cumplirse y que un diccionario no puede hacer respetar. Si solo estás agrupando tres datos que nadie va a modificar, un diccionario es más simple, se convierte a JSON solo y pandas lo entiende de fábrica.

Envolver todo en clases porque "así se programa bien" es el antipatrón más común de quien acaba de aprender POO. El tercer ejercicio del módulo trata exactamente de eso.

````python
# ---- Mini-lección: de la clase al objeto ----

class Cuenta:
    """Una cuenta con titular y saldo."""

    def __init__(self, titular, saldo=0.0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, monto):
        self.saldo += monto
        return self.saldo

    def __repr__(self):
        return f"Cuenta({self.titular!r}, {self.saldo})"


a = Cuenta("Ada")
b = Cuenta("Bo", 500.0)

print(a, "|", b)                       # gracias a __repr__
print("saldo inicial de Ada:", a.saldo)
a.depositar(100)
print("tras depositar      :", a.saldo, "| Bo sigue en", b.saldo, "<- son piezas independientes")

# El azucar sintactico, al descubierto: las dos lineas hacen LO MISMO
print(a.depositar(1))
print(Cuenta.depositar(a, 1))          # self = a, pasado a mano

# Lo que ya usabas sin saberlo: str tambien es una clase
print(type("hola").__name__, "|", "hola".upper(), "|", str.upper("hola"))
````

````python
# ---------- Verificación (no la edites) ----------
_pf = {"MX": 0.6, "BR": 0.4}
_rd = {"MX": 0.10, "BR": -0.05}

def _copia_protegida():
    """Mutar el diccionario original NO debe alterar un portafolio ya creado."""
    original = {"MX": 0.6, "BR": 0.4}
    p = Portafolio(original)
    original["MX"] = 99.0
    return p.pesos["MX"] == 0.6

probar("Ejercicio 110", [
    (lambda: round(Portafolio(_pf).rendimiento(_rd), 10), 0.04),
    (lambda: repr(Portafolio(_pf)), "Portafolio({'MX': 0.6, 'BR': 0.4})"),
    (lambda: tipo_de_error(Portafolio, {"MX": 0.6, "BR": 0.5}), "ValueError"),
    (lambda: tipo_de_error(Portafolio, {}), "ValueError"),
    (lambda: tipo_de_error(Portafolio, _pf), "sin error"),
    (lambda: round(Portafolio({"MX": 1.0}).rendimiento({"MX": 0.07}), 10), 0.07),
    (lambda: _copia_protegida(), True),
])
````

<summary>🔑 Solución del ejercicio 110 — <em>ábrela solo después de intentarlo</em></summary>

```python
class Portafolio:
    """Portafolio cuyos pesos suman 1 por construcción."""

    def __init__(self, pesos):
        if not pesos:
            raise ValueError("un portafolio necesita al menos un activo")
        if abs(sum(pesos.values()) - 1.0) > 1e-9:
            raise ValueError("los pesos deben sumar 1")
        self.pesos = dict(pesos)          # copia, para que nadie lo modifique por fuera

    def rendimiento(self, rendimientos):
        return sum(p * rendimientos[act] for act, p in self.pesos.items())

    def __repr__(self):
        return f"Portafolio({self.pesos})"


p = Portafolio({"MX": 0.6, "BR": 0.4})
print(p, "-> rendimiento:", round(p.rendimiento({"MX": 0.10, "BR": -0.05}), 4))

for malo in ({"MX": 0.6, "BR": 0.5}, {}):
    print(malo, "->", tipo_de_error(Portafolio, malo))
```

**Por qué:** Esto es lo que una clase compra y un diccionario no.

**El invariante se verifica una vez, en la frontera.** Con un diccionario suelto, *cada* función que lo recibe tendría que comprobar que los pesos suman 1 — o, peor, ninguna lo comprueba y el error aparece tres pasos después, cuando el rendimiento sale mal y nadie sabe por qué. Aquí el objeto **no puede existir en estado inválido**: si lo tienes en la mano, es válido. Eso es lo que significa encapsular.

**Falla temprano y con el error correcto.** `ValueError` es el adecuado: el tipo es bueno (un diccionario de números) y el valor es imposible. No es `assert`, porque esto valida una entrada esperable del usuario, no un supuesto interno — la distinción del ejercicio 96.

**`dict(pesos)` en lugar de `pesos`.** Sin la copia, quien te pasó el diccionario puede modificarlo después y romper tu invariante desde fuera, sin volver a pasar por `__init__`. Es el mismo bug de alias del ejercicio 64, ahora con consecuencias mayores porque violaría la garantía que la clase promete.

Nota lo que **no** hicimos: ni herencia, ni una clase base `Activo`, ni propiedades. Nada de eso hace falta para que la clase valga la pena. Lo que la justifica es una sola regla protegida.

</details>

## Módulo 12 — NumPy — cálculo vectorizado

**Ejercicios 112–119** · duración aproximada: 65–100 min

**Al terminar este módulo podrás:**

- Crear arrays y operar sin ciclos (vectorización)
- Filtrar con máscaras booleanas
- Entender `axis` y `shape`
- Usar broadcasting y resolver sistemas lineales
- Reacomodar y apilar arrays, y calcular cuantiles y atípicos

### Teoría

**NumPy** es la base numérica de todo el ecosistema científico de Python: pandas, scikit-learn, statsmodels y matplotlib están construidos encima.

Un `ndarray` es un bloque **contiguo** de memoria con un tipo único (`dtype`). Eso permite la **vectorización** (*vectorization*): la operación se aplica a todo el array en un ciclo escrito en C, típicamente 10–100× más rápido que un `for` de Python, y el código se lee como la fórmula matemática:

```python
r = (p[1:] - p[:-1]) / p[:-1]      # rendimientos de una serie de precios
z = (a - a.mean()) / a.std()       # estandarización
```

**Máscaras booleanas**: `a > a.mean()` produce un array de `True`/`False` y `a[mascara]` selecciona. Dentro de arrays se usan `&`, `|`, `~` (no `and`, `or`, `not`), y cada condición necesita paréntesis: `a[(a > 0) & (a < 10)]`.

**`shape` y `axis`**. `shape` es la forma: `(2, 3)` son 2 filas y 3 columnas. La regla que evita el 90% de las confusiones con `axis`: **`axis=k` significa que el eje `k` desaparece**. En una matriz `(2, 3)`, `mean(axis=0)` colapsa las 2 filas y deja 3 valores (uno por columna).

**Broadcasting**: NumPy estira automáticamente las formas compatibles sin copiar memoria, y por eso `a - a.mean()` funciona aunque uno sea array y el otro escalar.

**Álgebra lineal**: `@` es multiplicación matricial, `.T` transpone, `np.linalg.solve(A, b)` resuelve $A\mathbf{x}=\mathbf{b}$. Nunca resuelvas invirtiendo (`inv(A) @ b`): es más lento y numéricamente peor.

Una advertencia que ahorra horas: NumPy calcula la desviación estándar **poblacional** por defecto (`ddof=0`) y pandas la **muestral** (`ddof=1`). La misma columna dará números distintos según la librería.

````python
# ---- Mini-lección: NumPy ----
import numpy as np

precios = np.array([100., 102., 101., 105., 110.])
print(precios.shape, precios.dtype)

# Vectorización: sin un solo ciclo
rend = (precios[1:] - precios[:-1]) / precios[:-1]
print("rendimientos:", rend.round(4))
print("log-rend:    ", np.diff(np.log(precios)).round(4))

# Estadísticas
print(precios.mean(), precios.std(), precios.std(ddof=1), precios.max(), precios.argmax())

# Máscara booleana
print(precios > precios.mean())
print(precios[precios > precios.mean()])
print(precios[(precios > 100) & (precios < 106)])

# Matrices y axis
m = np.array([[1., 2., 3.], [4., 5., 6.]])
print(m.shape)
print("todo:      ", m.mean())
print("por columna", m.mean(axis=0))    # el eje 0 (filas) desaparece
print("por fila   ", m.mean(axis=1))
print("transpuesta\n", m.T)

# Broadcasting y álgebra lineal
print("centrada por columna\n", m - m.mean(axis=0))
A = np.array([[2., 1.], [1., 3.]])
b = np.array([8., 13.])
x = np.linalg.solve(A, b)
print("solución:", x, "| verificación:", A @ x)
````

> 🔮 **Predice antes de correr.** Para `m` de forma `(2,3)`: ¿qué forma tiene `m.mean(axis=0)`? ¿`(2,)` o `(3,)`?

```python
def rendimientos(precios):
    """Rendimientos simples periodo a periodo, sin ciclos."""
    p = np.array(precios, dtype=float)
    return (p[1:] - p[:-1]) / p[:-1]   # operación elemento a elemento sobre dos rebanadas

print(rendimientos([100, 110, 99]))

# Equivalente con np.diff, aún más directo:
def rendimientos_v2(precios):
    p = np.array(precios, dtype=float)
    return np.diff(p) / p[:-1]

# Rendimientos logarítmicos, los que se usan en finanzas por ser aditivos en el tiempo:
print(np.diff(np.log([100, 110, 99])))
```

**Por qué:** La **vectorización** es la razón de existir de NumPy: la operación corre en C sobre memoria contigua, típicamente 10–100× más rápido que un ciclo de Python, y el código se lee como la fórmula matemática. Los rendimientos logarítmicos $\ln(P_t/P_{t-1})$ se prefieren en finanzas porque se suman a lo largo del tiempo.

```python
    """Elementos por encima de la media."""
    a = np.array(valores, dtype=float)
    mascara = a > a.mean()      # array de booleanos, del mismo tamaño que a
    return a[mascara]           # selecciona donde la máscara es True

a = np.array([1., 2., 3., 4.])
print(a > a.mean())             # [False False  True  True]
print(sobre_promedio([1,2,3,4]))
print((a > a.mean()).sum())     # 2 -- sumar booleanos cuenta cuántos True hay
```

**Por qué:** La máscara booleana sustituye el `for` + `if` de Python puro y es la base del filtrado en pandas (`df[df.pib > 1]`). Un detalle importante: dentro de arrays se usan `&`, `|`, `~` (no `and`, `or`, `not`), y cada condición necesita paréntesis: `a[(a > 0) & (a < 10)]`.

```python
def por_eje(matriz):
    """Medias por columna (axis=0) y sumas por fila (axis=1)."""
    m = np.array(matriz, dtype=float)
    medias_col = m.mean(axis=0).tolist()   # colapsa el eje 0 -> un valor por columna
    sumas_fila = m.sum(axis=1).tolist()    # colapsa el eje 1 -> un valor por fila
    return medias_col, sumas_fila

m = np.array([[1,2,3],[4,5,6]], dtype=float)
print(m.shape)          # (2, 3): 2 filas, 3 columnas
print(m.mean())         # 3.5 -- sin axis, colapsa TODO
print(m.mean(axis=0))   # [2.5 3.5 4.5]
print(m.mean(axis=1))   # [2. 5.]
```

**Por qué:** La regla que desatasca el 90% de las confusiones: `axis=k` significa **"el eje k desaparece"**. En una matriz de forma `(2, 3)`, `mean(axis=0)` elimina la dimensión de 2 filas y deja 3 valores. Es la misma lógica que en pandas, donde `axis=0` es "por columnas" y `axis=1` es "por filas".

</details>

```python
def estandarizar(valores):
    """Puntajes z con desviación poblacional (ddof=0)."""
    a = np.array(valores, dtype=float)
    return ((a - a.mean()) / a.std()).tolist()

print(estandarizar([2,4,4,4,5,5,7,9]))

# Broadcasting también funciona entre formas compatibles:
m = np.array([[1., 2.], [3., 4.]])
print(m - m.mean(axis=0))   # resta la media de CADA columna
```

**Por qué:** El *broadcasting* permite operar arrays de formas distintas estirando la más pequeña sin copiar memoria. Ojo con `ddof`: NumPy usa por defecto la desviación **poblacional** ($n$) y pandas la **muestral** ($n-1$) — la misma columna te dará números ligeramente distintos según la librería, y eso ha arruinado más de una tabla de resultados.

```python
def resolver(A, b):
    """Solución de A x = b."""
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    return np.linalg.solve(A, b).tolist()

x = resolver([[2,1],[1,3]], [8,13])
print(x)   # [2.2, 3.6]

# Verificación: A @ x debe reproducir b
print(np.array([[2,1],[1,3]]) @ np.array(x))   # [ 8. 13.]

# Bonus: OLS por ecuaciones normales, beta = (X'X)^-1 X'y resuelto como sistema
def ols(X, y):
    X = np.array(X, dtype=float)
    y = np.array(y, dtype=float)
    return np.linalg.solve(X.T @ X, X.T @ y).tolist()
```

**Por qué:** `solve` usa factorización LU con pivoteo: $O(n^3)$ pero numéricamente estable. Invertir la matriz explícitamente cuesta más y amplifica el error de redondeo — en econometría, con matrices $X'X$ casi singulares por multicolinealidad, esa diferencia sí se ve en los coeficientes. El operador `@` es multiplicación matricial (Python 3.5+).

</details>

```python
def acomodar(valores, filas):
    """Convierte una lista plana en matriz de `filas` renglones."""
    return np.array(valores).reshape(filas, -1)   # -1 = deduce las columnas

def apilar(a, b):
    """Apila dos listas como renglones de una matriz."""
    return np.vstack([a, b])

print(acomodar(range(12), 3))
print(apilar([1,2,3], [4,5,6]))

# reshape NO copia datos: devuelve una VISTA sobre la misma memoria
a = np.arange(6)
b = a.reshape(2, 3)
b[0, 0] = 99
print(a)          # [99  1  2  3  4  5]  <- ¡cambió el original!
print(b.base is a)  # True: b comparte memoria con a

# Y el error cuando las dimensiones no cuadran:
print(tipo_de_error(acomodar, [1,2,3], 2))   # ValueError
```

**Por qué:** Dos cosas importantes. **`reshape` devuelve una vista, no una copia**: comparte memoria con el original, así que mutar una afecta a la otra — es lo que hace la operación casi gratuita, y también una fuente de bugs si no lo sabes (`.copy()` si quieres independencia). Y `hstack`/`vstack`/`concatenate` **sí** copian, porque tienen que construir un bloque contiguo nuevo. El `-1` en `reshape` es azúcar valiosa: `reshape(-1, 3)` significa "tres columnas, tantos renglones como hagan falta".

```python
def resumen_robusto(valores):
    """Estadísticos robustos basados en cuantiles."""
    a = np.array(valores, dtype=float)
    p25 = float(np.percentile(a, 25))
    p75 = float(np.percentile(a, 75))
    return {
        "mediana": float(np.median(a)),
        "p25": p25,
        "p75": p75,
        "iqr": p75 - p25,
        "orden_desc": np.argsort(a)[::-1].tolist(),   # índices, no valores
    }

d = [4, 8, 15, 16, 23, 42]
print(resumen_robusto(d))

# argsort da ÍNDICES: sirve para reordenar otra columna en paralelo
paises = np.array(["MX", "BR", "CL", "AR", "CO", "PE"])
idx = np.argsort(d)[::-1]
print(list(zip(paises[idx], np.array(d)[idx])))

# Detección de atípicos con la regla de Tukey (1.5 * IQR)
r = resumen_robusto(d)
limite = r["p75"] + 1.5 * r["iqr"]
print(f"límite superior: {limite} | atípicos: {[x for x in d if x > limite]}")
```

**Por qué:** Los cuantiles son **robustos**: un valor extremo mueve la media pero casi no mueve la mediana ni el IQR, y por eso la regla de Tukey ($Q_3 + 1.5\,\text{IQR}$) es el criterio estándar para marcar atípicos — es exactamente lo que dibujan los bigotes de un *boxplot*. Ojo con un detalle técnico: existen varios métodos de interpolación para calcular percentiles, y NumPy usa `linear` por defecto mientras otros programas usan otro; por eso el mismo Q1 puede diferir entre NumPy, Excel y Stata.

</details>

- Rankear, pivotar y evitar el error de `and` sobre una Series

### Teoría

**pandas** es NumPy con etiquetas. Dos estructuras:

- **`Series`**: un array **con índice**. El índice es lo que hace a pandas distinto: las operaciones se **alinean por etiqueta**, no por posición.
- **`DataFrame`**: una tabla; cada columna es una `Series` y todas comparten el índice de renglones.

```python
df = pd.DataFrame({"pais": [...], "pib": [...]})
df["nueva"] = df["pib"] / df["poblacion"]     # columna calculada, vectorizada
df[df["pib"] > 1000]                          # filtrado booleano
df.groupby("region")["pib"].mean()             # split-apply-combine
pd.merge(a, b, on="pais", how="left")          # unión de tablas
```

Métodos de reconocimiento inicial de cualquier tabla: `.head()`, `.info()`, `.describe()`, `.shape`, `.columns`, `.dtypes`, `.isna().sum()`.

Tres trampas que valen la advertencia:

1. **`and` no funciona sobre Series.** Hay que usar `&` y `|`, con paréntesis: `df[(df.a > 1) & (df.b < 2)]`. Con `and` obtienes `ValueError: The truth value of a Series is ambiguous`.
2. **`NaN` no es cero.** `NaN` (*Not a Number*) representa ausencia. Las agregaciones lo ignoran por defecto, pero rellenarlo con `.fillna(0)` es una **decisión analítica**, no una limpieza neutra: una inflación desconocida no es 0%.
3. **Nunca iteres renglón por renglón** con `for i in range(len(df))`. Toda operación por columnas es vectorizada y cientos de veces más rápida.

El tipo de unión también es una decisión: `left` conserva tu universo de referencia, `inner` solo las coincidencias, `outer` todo.

````python
# ---- Mini-lección: pandas ----
import pandas as pd

df = pd.DataFrame({
    "pais": ["MX", "BR", "CL", "AR", "CO"],
    "region": ["Norteamerica", "Sudamerica", "Sudamerica", "Sudamerica", "Sudamerica"],
    "pib": [1790.0, 2170.0, 340.0, 640.0, 415.0],       # miles de millones USD
    "poblacion": [129.0, 216.0, 19.6, 46.0, 52.0],      # millones
    "inflacion": [4.7, 4.6, 3.9, 211.4, 6.2],
})

print(df.shape, list(df.columns))
print(df.head(3))
print(df.describe().round(2))

# Columna calculada (vectorizada)
df["pib_per_capita"] = df["pib"] / df["poblacion"] * 1000
print(df[["pais", "pib_per_capita"]].round(0))

# Filtrado booleano: & y | con paréntesis, nunca and/or
print(df[(df["pib"] > 500) & (df["inflacion"] < 10)][["pais", "pib", "inflacion"]])

# Orden y ranking
print(df.sort_values("pib_per_capita", ascending=False)[["pais", "pib_per_capita"]].round(0))

# groupby: split-apply-combine
print(df.groupby("region")["pib"].agg(["count", "sum", "mean"]).round(1))

# merge y NaN
extra = pd.DataFrame({"pais": ["MX", "BR"], "desempleo": [2.7, 7.8]})
unido = pd.merge(df, extra, on="pais", how="left")
print(unido[["pais", "desempleo"]])
print("faltantes por columna:\n", unido.isna().sum())
````

```python
def unir(df_pib, df_inflacion):
    """Unión por la izquierda con faltantes rellenados en cero."""
    df = pd.merge(df_pib, df_inflacion, on="pais", how="left")
    df["inflacion"] = df["inflacion"].fillna(0.0)
    return df

pib_demo = pd.DataFrame({"pais": ["MX", "BR", "CL"], "pib": [1790.0, 2170.0, 340.0]})
inf_demo = pd.DataFrame({"pais": ["MX", "BR"], "inflacion": [4.7, 4.6]})

# Antes de rellenar, así se ve el faltante:
print(pd.merge(pib_demo, inf_demo, on="pais", how="left"))
print(unir(pib_demo, inf_demo))
```

**Por qué:** El tipo de unión es una decisión analítica: `left` conserva tu universo de referencia, `inner` se queda solo con las coincidencias, `outer` con todo. Y rellenar `NaN` con `0.0` **no es neutro**: una inflación desconocida no es cero. A veces lo correcto es dejar el `NaN` y que las agregaciones lo excluyan explícitamente.

</details>

```python
def con_metricas(df):
    """Agrega clave, ranking y participación porcentual."""
    df = df.copy()                                   # evita SettingWithCopyWarning y mutar al llamador
    df["clave"] = df["pais"].apply(lambda p: p.upper())
    df["rank"] = df["pib"].rank(ascending=False, method="min").astype(int)
    df["participacion"] = (df["pib"] / df["pib"].sum() * 100).round(2)
    return df

d = pd.DataFrame({"pais": ["mx","br","cl"], "pib": [1790.0, 2170.0, 340.0]})
print(con_metricas(d))

# .apply() con lambda es genérico pero lento; para texto hay métodos vectorizados:
print(d["pais"].str.upper().tolist())      # más rápido que apply(lambda p: p.upper())

# method= cambia el desempate del ranking:
s = pd.Series([10, 10, 5])
for m in ("min", "max", "average", "dense"):
    print(m, s.rank(ascending=False, method=m).tolist())
```

**Por qué:** El `df = df.copy()` de la primera línea no es paranoia: sin él, la función modifica el DataFrame del llamador (mismo problema de alias del ejercicio de listas) y en pandas puede además disparar el temido `SettingWithCopyWarning`. Sobre `method` en `rank`: `"min"` da 1,1,3 a un empate (como en deportes), `"average"` da 1.5,1.5,3 y `"dense"` da 1,1,2 — elegir mal cambia tu tabla de posiciones sin avisar.

</details>

```python
def pivote(df):
    """Formato largo -> ancho: regiones en renglones, años en columnas."""
    return df.pivot_table(index="region", columns="anio", values="pib", aggfunc="sum")

demo = pd.DataFrame({
    "region": ["Norte", "Sur", "Sur", "Norte"],
    "anio": [2023, 2023, 2024, 2024],
    "pib": [100.0, 200.0, 300.0, 400.0],
})
tabla = pivote(demo)
print(tabla)

# El camino de regreso (ancho -> largo) es melt:
print(tabla.reset_index().melt(id_vars="region", value_name="pib"))

# Con datos faltantes aparecen NaN, y puedes rellenarlos al pivotar:
print(demo.iloc[:3].pivot_table(index="region", columns="anio", values="pib",
                                aggfunc="sum", fill_value=0))
```

**Por qué:** Formato **largo** (una observación por renglón) contra **ancho** (una columna por periodo) es la decisión de estructura más frecuente al analizar datos: el largo es mejor para calcular y graficar, el ancho para presentar y para que un humano lo lea. `pivot_table` va de largo a ancho y `melt` regresa; en el ecosistema de R esto se llama *tidy data* y es el mismo concepto. Un detalle: `pivot_table` **agrega** (por eso pide `aggfunc`), mientras `pivot` falla si hay duplicados.

#### Ejercicio 127 — 🐛 Depura: `and` no funciona sobre una Series

**Concepto:** *Ambiguous truth value* en pandas

Este filtro se ve razonable y no corre:

```python
def filtrar_mal(df, minimo, maximo):
    return df[df["pib"] > minimo and df["pib"] < maximo]
```

Lanza `ValueError: The truth value of a Series is ambiguous`. **Diagnostica:** `and` necesita convertir cada lado a un solo `True`/`False`, pero cada lado es una Series de varios booleanos — ¿debería ser verdadera si *todos* lo son, o si *alguno* lo es? pandas se niega a adivinar.

Usa `&` en lugar de `and`, y **paréntesis obligatorios** en cada condición: `(df["pib"] > minimo) & (df["pib"] < maximo)`. Sin los paréntesis, `&` tiene mayor precedencia que `>` y todo se rompe.

</details>

````python
# ---------- Ejercicio 127: TU SOLUCIÓN ----------
def filtrar_mal(df, minimo, maximo):
    """NO la modifiques: es el bug que estamos estudiando."""
    return df[df["pib"] > minimo and df["pib"] < maximo]

def filtrar(df, minimo, maximo):
    pass
````

````python
# ---------- Verificación (no la edites) ----------
_dfa = pd.DataFrame({"pais": ["MX","BR","CL","AR"], "pib": [1790.0, 2170.0, 340.0, 640.0]})
probar("Ejercicio 127", [
    (lambda: tipo_de_error(filtrar_mal, _dfa, 500, 2000), "ValueError"),
    (lambda: filtrar(_dfa, 500, 2000)["pais"].tolist(), ["MX", "AR"]),
    (lambda: filtrar(_dfa, 0, 10000)["pais"].tolist(), ["MX", "BR", "CL", "AR"]),
    (lambda: filtrar(_dfa, 5000, 9000)["pais"].tolist(), []),
    (lambda: filtrar(_dfa, 500, 2000).shape, (2, 2)),
])
````

<details>
<summary>🔑 Solución del ejercicio 127 — <em>ábrela solo después de intentarlo</em></summary>

```python
def filtrar_mal(df, minimo, maximo):
    return df[df["pib"] > minimo and df["pib"] < maximo]

def filtrar(df, minimo, maximo):
    """Renglones con pib estrictamente entre los dos límites."""
    return df[(df["pib"] > minimo) & (df["pib"] < maximo)]

demo = pd.DataFrame({"pais": ["MX","BR","CL","AR"], "pib": [1790.0, 2170.0, 340.0, 640.0]})
print(tipo_de_error(filtrar_mal, demo, 500, 2000))   # ValueError, no un resultado equivocado
print(filtrar(demo, 500, 2000))

# Alternativas legibles cuando hay varias condiciones:
print(demo.query("500 < pib < 2000"))                      # query acepta sintaxis de Python
print(demo[demo["pib"].between(500, 2000, inclusive="neither")])

# Por qué falla and: cada lado es una Series de booleanos, no un booleano
mascara = demo["pib"] > 500
print(mascara.tolist(), "| .all():", mascara.all(), "| .any():", mascara.any())
```

**Por qué:** `and` y `or` exigen un valor de verdad único, y una Series de cuatro booleanos no lo tiene; pandas lanza el error en lugar de elegir por ti (a diferencia de NumPy en versiones antiguas, que a veces adivinaba). Los operadores correctos son `&`, `|` y `~`, que operan **elemento a elemento**, y necesitan paréntesis porque su precedencia es mayor que la de los comparadores. Cuando hay tres o más condiciones, `df.query("...")` suele quedar más legible que una cadena de `&`.

</details>

### Teoría

Hasta aquí, todos los datos del curso venían escritos a mano dentro del código. Eso es cómodo para aprender y no se parece en nada al trabajo real, donde los datos llegan en un archivo que alguien más armó, con las convenciones de alguien más y los errores de alguien más.

**Archivos.** La forma correcta de abrirlos es con un **gestor de contexto** (*context manager*):

```python
with open(ruta, "w", encoding="utf-8") as f:
    f.write("texto\n")
# al salir del bloque, el archivo se cierra solo, incluso si hubo una excepción
```

Los modos son `"r"` (leer), `"w"` (escribir, **borra lo que había**) y `"a"` (agregar al final). Escribe siempre `encoding="utf-8"` de forma explícita: en Windows el default es otro y los acentos se corrompen.

**CSV con pandas.** `df.to_csv(ruta, index=False)` y `pd.read_csv(ruta)`. Un CSV es texto plano y **no guarda tipos**: pandas los infiere al leer, con acierto casi siempre. Los archivos reales necesitan argumentos:

| Argumento | Para qué |
|---|---|
| `sep=";"` | separador que no es coma (común en México y Europa) |
| `decimal=","` | coma decimal |
| `thousands="."` | separador de miles |
| `na_values=["n/d", "ND", "-"]` | qué textos significan "faltante" |
| `encoding="latin-1"` | archivos viejos de Windows |
| `skiprows=3` | archivos que traen títulos antes de la tabla |
| `dtype={"cp": str}` | evitar que un código postal se vuelva número |

**JSON.** Es el formato de las APIs: `json.loads(texto)` lo convierte en diccionarios y listas de Python (`null` se vuelve `None`). Casi siempre viene **anidado**, y desanidarlo es trabajo de ciclos y `.get()`.

**HTTP.** `requests.get(url, timeout=5)` descarga. El `timeout` no es opcional: sin él, una petición puede colgar tu notebook indefinidamente. Y como la red falla de docenas de formas distintas, aquí sí se justifica un `except Exception` amplio — con la condición de devolver algo que el llamador **deba** verificar.

**Gráficas.** matplotlib tiene dos interfaces; usa la orientada a objetos:

```python
fig, ax = plt.subplots(figsize=(7, 4))   # fig = lienzo, ax = ejes
ax.bar(x, y); ax.set_title("...")
fig.tight_layout(); fig.savefig("salida.png", dpi=150)
```

Una gráfica no se puede verificar con `assert`, pero su estructura sí: título, número de barras, existencia del archivo. Eso es lo que revisan las pruebas de este módulo.

```python
def guardar_y_leer(ruta, lineas):
    """Escribe una lista de textos como renglones y la vuelve a leer."""
    with open(ruta, "w", encoding="utf-8") as f:      # "w" sobreescribe; "a" agrega
        for linea in lineas:
            f.write(linea + "\n")
    # al salir del with, el archivo ya está cerrado y escrito en disco
    with open(ruta, "r", encoding="utf-8") as f:
        return f.read().splitlines()

print(guardar_y_leer("prueba_curso.txt", ["MX,4.7", "BR,4.6", "Querétaro"]))

# Los tres modos que usarás: leer, sobreescribir, agregar
with open("prueba_curso.txt", "a", encoding="utf-8") as f:
    f.write("CL,3.9\n")
with open("prueba_curso.txt", encoding="utf-8") as f:
    print(f.read())

# En Colab los archivos viven en /content y se BORRAN al reiniciar el entorno.
import os
print(os.getcwd(), "|", [a for a in os.listdir() if a.endswith(".txt")])
```

**Por qué:** El `with` es un **gestor de contexto** (*context manager*): garantiza que el archivo se cierre incluso si ocurre una excepción a la mitad. Sin él, un archivo abierto en modo escritura puede quedarse con datos en el búfer y perderlos. Dos detalles que muerden en México: `encoding="utf-8"` explícito (en Windows el default es `cp1252` y los acentos se corrompen), y que el modo `"w"` **borra el archivo sin preguntar** — el número de datos perdidos por confundir `"w"` con `"a"` es difícil de exagerar.

</details>

```python
    return pd.read_csv(ruta)
```

```python
import io

def leer_sucio(texto):
    """Lee un CSV con convenciones latinoamericanas y encabezados sucios."""
    df = pd.read_csv(
        io.StringIO(texto),   # trata un texto en memoria como si fuera un archivo
```

```python
import json

def extraer(texto_json):
    """Desanida la respuesta de la API del Banco Mundial."""
    datos = json.loads(texto_json)
    registros = datos[1]                       # [0] es metadata, [1] son los datos
    salida = []
    for r in registros:
        if r.get("value") is None:             # omite los faltantes
            continue
        salida.append((r["country"]["value"], int(r["date"]), r["value"]))
    return salida

crudo = json.dumps([{"page": 1}, [
    {"country": {"value": "Mexico"}, "date": "2024", "value": 4.72},
    {"country": {"value": "Brazil"}, "date": "2024", "value": None},
]])
print(extraer(crudo))

# De lista de tuplas a DataFrame en una línea:
df = pd.DataFrame(extraer(crudo), columns=["pais", "anio", "inflacion"])
print(df)

# pandas también sabe desanidar solo, si la estructura es regular:
registros = json.loads(crudo)[1]
print(pd.json_normalize(registros)[["country.value", "date", "value"]])
```

**Por qué:** El patrón `r["country"]["value"]` es el que se rompe en producción: si un registro no trae `country`, obtienes `KeyError` y el proceso muere a medio camino. Por eso se usa `.get()` con default en los campos opcionales (como aquí con `value`) y por eso el filtro de `None` va **antes** de tocar los datos. `pd.json_normalize` automatiza el desanidado cuando la estructura es regular, pero cuando no lo es —el caso común— el ciclo explícito es más robusto y mucho más fácil de depurar.

</details>

```python
import requests

def descargar_json(url, timeout=5):
    """Descarga JSON con tolerancia total a fallas de red. None si algo sale mal."""
    try:
        r = requests.get(url, timeout=timeout)   # timeout SIEMPRE
        if r.status_code != 200:
            return None                          # 404, 500, 403...
        return r.json()                          # lanza si el cuerpo no es JSON
    except Exception:                            # ConnectionError, Timeout, JSONDecodeError...
        return None

print(descargar_json("http://no-existe-jamas.invalid/x", timeout=3))   # None

# La llamada real a la API del Banco Mundial (inflación anual de México y Brasil).
# Si tienes internet verás los datos; si no, verás el aviso y el programa NO se cae.
URL = ("https://api.worldbank.org/v2/country/MX;BR/indicator/FP.CPI.TOTL.ZG"
       "?format=json&date=2015:2024&per_page=100")

datos = descargar_json(URL, timeout=10)
if datos is None:
    print("Sin conexión o API no disponible: seguimos con los datos de ejemplo.")
else:
    filas = [(r["country"]["value"], int(r["date"]), r["value"])
             for r in datos[1] if r.get("value") is not None]
    df = pd.DataFrame(filas, columns=["pais", "anio", "inflacion"])
    print(df.pivot_table(index="anio", columns="pais", values="inflacion").round(2))
```

**Por qué:** Tres decisiones de diseño aquí. **El `timeout` explícito:** sin él, `requests` espera para siempre y tu notebook queda colgado — es el error más común al consumir APIs. **El `except Exception` amplio:** justificable porque todas las fallas de red son equivalentes para el llamador, pero nota que devolvemos `None` y quien llama **debe** verificarlo (como hace el `if datos is None`). **El diseño degradable:** el código sigue funcionando sin internet, con datos de ejemplo, en lugar de reventar — indispensable para material de clase, donde el wifi del aula es una variable aleatoria.

</details>

```python
import matplotlib.pyplot as plt

def grafica_pib(df, ruta="pib.png"):
    """Gráfica de barras del PIB, guardada en disco. Devuelve los ejes."""
    fig, ax = plt.subplots(figsize=(7, 4))          # fig = lienzo, ax = los ejes
    ax.bar(df["pais"], df["pib"], color="#4C72B0")
    ax.set_title("PIB por país")
    ax.set_ylabel("Miles de millones de USD")
    ax.spines[["top", "right"]].set_visible(False)  # menos tinta, más datos
    for i, v in enumerate(df["pib"]):               # etiquetas de valor
        ax.text(i, v, f"{v:,.0f}", ha="center", va="bottom", fontsize=9)
    fig.tight_layout()
    fig.savefig(ruta, dpi=150)
    return ax

demo = pd.DataFrame({"pais": ["MX", "BR", "CL"], "pib": [1790.0, 2170.0, 340.0]})
ax = grafica_pib(demo, "pib_demo.png")
plt.show()

# Serie de tiempo con dos líneas, el otro gráfico que usarás todo el tiempo:
anios = list(range(2015, 2025))
mx = [2.7, 2.8, 6.0, 4.9, 3.6, 3.4, 5.7, 7.9, 5.5, 4.7]
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(anios, mx, marker="o", label="México")
ax.axhline(3.0, linestyle="--", linewidth=1, color="gray", label="Meta Banxico 3%")
ax.set_title("Inflación anual"); ax.set_ylabel("%"); ax.legend()
fig.tight_layout(); plt.show()
```

**Por qué:** matplotlib tiene dos interfaces y conviene elegir una: la de estado global (`plt.bar`, `plt.title`) es cómoda para un gráfico rápido, y la **orientada a objetos** (`fig, ax = plt.subplots()`) es la que se usa en código serio, porque hace explícito sobre qué ejes estás dibujando y permite varias gráficas en una figura. Devolver `ax` en lugar de solo guardar el archivo es lo que hace la función **composable**: quien la llama puede seguir modificando la gráfica. Y `fig.tight_layout()` evita que las etiquetas se corten, que es el defecto más visible de una gráfica hecha con prisa.

</details>

### Teoría

El cierre del curso: implementar tres estadísticos **desde su definición**, y después verificarlos contra la librería. Esa doble vía es el hábito profesional — si tu fórmula a mano y NumPy coinciden hasta el octavo decimal, entiendes lo que estás calculando; si solo llamas a la librería, no lo sabes.

**Correlación de Pearson.** Con $d_{xi} = x_i - \bar{x}$:

$$r = \frac{\sum d_{xi} d_{yi}}{\sqrt{\sum d_{xi}^2}\sqrt{\sum d_{yi}^2}} \in [-1, 1]$$

Los factores $1/n$ se cancelan, por eso basta con las sumas. Mide asociación **lineal**: $r \approx 0$ no implica independencia.

**Índice de Gini.** Sobre datos ordenados de menor a mayor:

$$G = \frac{2\sum_{i=1}^n i\,x_{(i)}}{n\sum_i x_i} - \frac{n+1}{n}$$

Es el doble del área entre la curva de Lorenz y la línea de igualdad perfecta. $G=0$ es igualdad total; el máximo alcanzable con $n$ datos es $1 - 1/n$.

**Mínimos cuadrados ordinarios (MCO / OLS).** Para $y = \beta_0 + \beta_1 x + u$:

$$\hat{\beta}_1 = \frac{S_{xy}}{S_{xx}} = \frac{\widehat{\text{Cov}}(x,y)}{\widehat{\text{Var}}(x)}, \qquad \hat{\beta}_0 = \bar{y} - \hat{\beta}_1\bar{x}$$

La pendiente es literalmente covarianza sobre varianza, de donde sale $\hat\beta_1 = r\,s_y/s_x$. La condición $S_{xx} > 0$ es la versión mínima del supuesto de rango completo: sin variación en el regresor, el parámetro no está identificado.

**Del estadístico a la decisión.** Un estadístico por sí solo no decide nada: hay que traducirlo a una probabilidad. El **valor p** es la probabilidad de observar algo al menos tan extremo como lo que viste *si la hipótesis nula fuera cierta*, y bajo la normal estándar se calcula como $p = 2\Phi(-|z|)$. Los dos últimos ejercicios del módulo cierran ese círculo con `statistics.NormalDist`, que viene en la librería estándar: no hace falta instalar nada.

El **proyecto integrador** del final junta todo en el patrón que sigue cualquier análisis serio: **validar → limpiar → transformar → agregar → reportar**, devolviendo datos (un diccionario) en lugar de imprimir, para que el resultado se pueda encadenar, probar y automatizar.

````python
# ---- Mini-lección: verificar tus fórmulas contra la librería ----
import numpy as np

x = [1., 2., 3., 4., 5.]
y = [2.1, 3.9, 6.2, 7.8, 10.1]

# Correlación: a mano vs NumPy
n = len(x)
mx, my = sum(x)/n, sum(y)/n
sxy = sum((xi-mx)*(yi-my) for xi, yi in zip(x, y))
sxx = sum((xi-mx)**2 for xi in x)
syy = sum((yi-my)**2 for yi in y)
r_manual = sxy / (sxx**0.5 * syy**0.5)
print(f"r manual: {r_manual:.10f}")
print(f"r numpy : {np.corrcoef(x, y)[0,1]:.10f}")

# MCO: a mano vs polyfit
b1 = sxy / sxx
b0 = my - b1*mx
print(f"manual  -> b0={b0:.6f}  b1={b1:.6f}")
pend, inter = np.polyfit(x, y, 1)      # ¡polyfit devuelve pendiente primero!
print(f"polyfit -> b0={inter:.6f}  b1={pend:.6f}")

# Relación entre ambos: b1 = r * sy/sx
sy = (syy/(n-1))**0.5
sx = (sxx/(n-1))**0.5
print(f"r * sy/sx = {r_manual * sy/sx:.6f}  (debe igualar b1)")

# R^2 en regresión simple es exactamente r^2
y_hat = [b0 + b1*xi for xi in x]
ss_res = sum((yi-yh)**2 for yi, yh in zip(y, y_hat))
print(f"R2 = {1 - ss_res/syy:.10f}   r^2 = {r_manual**2:.10f}")
````

> 🔮 **Predice antes de correr.** En regresión simple, ¿esperas que $R^2$ sea exactamente igual a $r^2$? La celda lo comprueba con 10 decimales.

```python
import math

def correlacion(x, y):
    """Coeficiente de correlación de Pearson, implementado desde la definición."""
    n = len(x)
    if n < 2 or n != len(y):
        return None
    mx = sum(x) / n
    my = sum(y) / n
    s_xy = s_xx = s_yy = 0.0
    for xi, yi in zip(x, y):
        dx, dy = xi - mx, yi - my
        s_xy += dx * dy      # covarianza (sin dividir)
        s_xx += dx * dx      # varianza de x (sin dividir)
        s_yy += dy * dy
    if s_xx == 0 or s_yy == 0:   # serie constante: correlación indefinida
        return None
    return s_xy / math.sqrt(s_xx * s_yy)

print(correlacion([1,2,3,4], [2,4,5,8]))
print(np.corrcoef([1,2,3,4], [2,4,5,8])[0,1])   # debe coincidir
```

**Por qué:** Los factores $1/n$ se cancelan entre numerador y denominador, por eso basta con las sumas de desviaciones. Nota que $r$ mide **asociación lineal**: $r \approx 0$ no significa independencia (piensa en $y = x^2$ centrada en cero), y $r$ es invariante a cambios de escala y origen — es exactamente la covarianza de las variables estandarizadas.

```python
def gini(valores):
    """Índice de Gini: 0 = igualdad perfecta, ->1 = concentración total."""
    if not valores:
        return None
    orden = sorted(valores)
    n = len(orden)
    total = sum(orden)
    if total == 0:
        return None
    suma_ponderada = sum(i * x for i, x in enumerate(orden, start=1))
    return (2 * suma_ponderada) / (n * total) - (n + 1) / n

for caso in ([1,1,1,1], [1,2,3,4], [0,0,0,4]):
    print(caso, "->", round(gini(caso), 4))
```

**Por qué:** Esta fórmula es la de la curva de Lorenz discretizada: $G$ es el doble del área entre la línea de igualdad y la curva. Con muestras pequeñas el estimador está sesgado a la baja (el máximo alcanzable es $1 - 1/n$, no 1), así que a veces se aplica la corrección $G \cdot n/(n-1)$. El Gini del ingreso en México ronda 0.42–0.45 según la ENIGH.

```python
def ols_simple(x, y):
    """Estimadores MCO de la regresión simple y = b0 + b1 x."""
    n = len(x)
    if n < 2 or n != len(y):
        return None
    mx = sum(x) / n
    my = sum(y) / n
    s_xy = s_xx = 0.0
    for xi, yi in zip(x, y):
        s_xy += (xi - mx) * (yi - my)
        s_xx += (xi - mx) ** 2
    if s_xx == 0:            # x sin variación: la pendiente no está identificada
        return None
    beta1 = s_xy / s_xx
    beta0 = my - beta1 * mx
    return beta0, beta1

print(ols_simple([1,2,3,4], [2,4,5,8]))          # (0.0, 1.9)
print(np.polyfit([1,2,3,4], [2,4,5,8], 1))       # [1.9 0.0] -> polyfit devuelve [pendiente, intercepto]
```

**Por qué:** $\hat{\beta}_1 = S_{xy}/S_{xx} = \widehat{\text{Cov}}(x,y)/\widehat{\text{Var}}(x)$: la pendiente es literalmente covarianza sobre varianza, y de ahí sale la relación $\hat\beta_1 = r \cdot s_y/s_x$. El requisito $S_{xx} > 0$ es la versión más simple de la condición de rango completo: sin variación en el regresor, el parámetro no está identificado. Nota que `np.polyfit` devuelve los coeficientes en **orden inverso** (grado descendente).

</details>

```python
import math

def prueba_t(x, y):
    """Prueba t de Welch. Devuelve (estadístico t, grados de libertad)."""
    nx, ny = len(x), len(y)
    if nx < 2 or ny < 2:
        return None

    s2x = varianza_muestral(x)      # función del módulo de funciones II
    s2y = varianza_muestral(y)
    a, b = s2x / nx, s2y / ny       # varianzas de las medias muestrales

    if a + b == 0:                  # ambas muestras constantes e iguales
        return 0.0, float(nx + ny - 2)

    t = (sum(x) / nx - sum(y) / ny) / math.sqrt(a + b)
    gl = (a + b) ** 2 / (a ** 2 / (nx - 1) + b ** 2 / (ny - 1))
    return t, gl

t, gl = prueba_t([10, 12, 14], [20, 22, 24])
print(f"t = {t:.4f}, gl = {gl:.2f}")
print("|t| > 2 sugiere diferencia significativa al 5% con gl moderados")

# Contraste con scipy, si está disponible en tu entorno:
try:
    from scipy import stats
    print(stats.ttest_ind([10,12,14], [20,22,24], equal_var=False))
except ImportError:
    print("scipy no está instalado; tu implementación es la referencia")
```

**Por qué:** La prueba de Welch es la que deberías usar por defecto: no supone varianzas iguales y su costo cuando sí lo son es despreciable, así que la vieja prueba t "agrupada" prácticamente no tiene caso. Nota lo que significa el estadístico: la diferencia de medias **medida en unidades de su propio error estándar**. Con $|t|$ grande la diferencia es grande *relativa al ruido*, no en términos absolutos — y con muestras de tamaño 3 como estas, el valor p asociado es poco confiable aunque $|t|$ salga alto. Significancia estadística no es relevancia económica.

</details>

```python
import math

def ic_media(datos, z=1.96):
    """Intervalo de confianza para la media, basado en la normal."""
    n = len(datos)
    if n < 2:
        return None
    m = sum(datos) / n
    s = desviacion_estandar(datos)        # muestral, con n-1
    error_estandar = s / math.sqrt(n)
    margen = z * error_estandar
    return m - margen, m + margen

datos = [2, 4, 4, 4, 5, 5, 7, 9]
inf, sup = ic_media(datos)
print(f"media = {sum(datos)/len(datos):.2f}  IC95% = [{inf:.4f}, {sup:.4f}]")

# El intervalo se angosta con la raíz de n: cuadruplicar la muestra lo parte a la mitad
for k in (1, 4, 16, 64):
    muestra = datos * k
    i, s_ = ic_media(muestra)
    print(f"n = {len(muestra):>4}  ancho = {s_ - i:.4f}")
```

**Por qué:** Dos precisiones que importan. **Interpretación:** el intervalo no dice "hay 95% de probabilidad de que la media verdadera esté aquí" —esa es la lectura bayesiana, con otro aparato— sino que si repitieras el muestreo muchas veces, el 95% de los intervalos construidos así contendrían el parámetro. **Método:** con $n$ pequeña deberías usar el cuantil de la $t$ de Student con $n-1$ grados de libertad en lugar de 1.96, porque estimaste $\sigma$ con los mismos datos; el 1.96 solo es apropiado con $n$ grande. Y el último ciclo muestra el rendimiento decreciente de muestrear más: el ancho cae con $\sqrt{n}$, así que reducir el error a la mitad exige **cuadruplicar** la muestra.

</details>

```python
from statistics import NormalDist

def valor_p(z):
    """Valor p de dos colas bajo la normal estándar."""
    return 2 * NormalDist().cdf(-abs(z))

for z in (0, 1, 1.96, 2.58, 3):
    print(f"z = {z:>5} -> p = {valor_p(z):.6f}")
```

**Por qué:** Tres cosas que casi nunca se dicen juntas.

**Qué es.** El valor p mide la rareza de tus datos **bajo la nula**, no la probabilidad de que la nula sea cierta. `p = 0.03` no significa "hay 3% de probabilidad de que no haya efecto": significa "si no hubiera efecto, datos así de extremos aparecerían el 3% de las veces". La confusión entre ambas es probablemente el error más repetido en la literatura aplicada.

**Por qué la normal y no la t.** Estás aproximando: el estadístico de Welch sigue una $t$ con los grados de libertad que calculaste, no una normal. Con $gl$ grandes ($>30$) la diferencia es despreciable; con $gl$ pequeños la normal te da un valor p **demasiado optimista**, es decir, te hace rechazar de más. La versión correcta necesita la CDF de la $t$, que no está en la librería estándar: `scipy.stats.t.sf(abs(t), gl) * 2`. Este curso se queda en la normal a propósito, para no depender de scipy — pero ahora sabes exactamente qué estás sacrificando y cuándo importa.

**De dónde salía el "|t| > 2".** De aquí: $p(1.96) \approx 0.05$. La regla de pulgar era esta cuenta, redondeada.

</details>

```python
def contraste(x, y, alfa=0.05):
    """Prueba t de Welch completa: estadístico, gl, valor p y decisión."""
    resultado = prueba_t(x, y)          # ejercicio 137
    if resultado is None:
        return None
    t, gl = resultado
    p = valor_p(t)                      # ejercicio 139
    return {"t": t, "gl": gl, "p": p, "rechaza_h0": p < alfa}

for nombre, a, b in [("distintos", [12,15,14,16,13,15,17,14,16,15],
                                   [18,20,19,21,17,20,22,19,21,20]),
                     ("parecidos", [10,11,12,11,10], [11,10,12,12,11])]:
    r = contraste(a, b)
    print(f"{nombre:>10}: t={r['t']:+.3f}  gl={r['gl']:.1f}  p={r['p']:.4f}  rechaza={r['rechaza_h0']}")
```

**Por qué:** Ya puedes hacer un contraste de hipótesis de principio a fin. Cuatro advertencias que valen más que la función.

**`alfa` es una elección tuya, no un descubrimiento.** El último caso de la verificación lo deja incómodamente claro: con `alfa=0.9` el mismo par de muestras "parecidas" pasa a rechazarse. El 0.05 es una convención heredada de Fisher, no una propiedad de la naturaleza. Decídelo **antes** de ver los datos.

**Devolver un diccionario y no imprimir** es lo que permite lo que sigue: guardar el resultado, compararlo con el del mes pasado, correr el contraste sobre cien pares de series. Un `print` no se puede encadenar.

**Y justo por eso, cuidado con correrlo cien veces.** Si pruebas 100 pares sin efecto real con `alfa=0.05`, esperas ~5 rechazos por puro azar. Eso es el *problema de comparaciones múltiples*, y es la razón por la que "busqué en los datos hasta que algo salió significativo" no es un método. Corregirlo (Bonferroni, Holm, FDR) queda fuera de este curso, pero saber que el problema existe es lo que separa usar la herramienta de que la herramienta te use.

**Rechazar no es probar.** `rechaza_h0 = True` dice que los datos son poco compatibles con la nula; no dice que tu hipótesis alternativa sea correcta, ni cuán grande es el efecto. Para eso está el intervalo de confianza del ejercicio 138, que sí habla en las unidades del problema.

</details>

## Módulo 16 — Series de tiempo

**Ejercicios 141–146** · duración aproximada: 70–110 min

**Al terminar este módulo podrás:**

- Convertir fechas en texto a un índice temporal ordenado
- Calcular rendimientos simples y logarítmicos, y saber cuándo usar cada uno
- Resumir con ventanas móviles y cambiar la frecuencia de una serie
- Alinear dos series con calendarios distintos sin inventar datos
- Reconocer y corregir el sesgo de anticipación, el error que arruina backtests

### Teoría

Todo lo que hiciste hasta aquí trata las filas como intercambiables: da igual el orden en que llegue el PIB de México y el de Brasil. Una **serie de tiempo** rompe ese supuesto — el orden *es* el dato — y con él se rompen varias costumbres.

**El índice es la fecha.** `pd.to_datetime` convierte texto a marcas temporales y `set_index` las vuelve el índice. A partir de ahí pandas alinea por fecha automáticamente en cualquier operación entre series, que es la mitad del valor de usarlo.

**Rendimientos.** Con precios $P_t$, el rendimiento simple y el logarítmico son

$$r_t = \frac{P_t - P_{t-1}}{P_{t-1}}, \qquad \ell_t = \ln\frac{P_t}{P_{t-1}}$$

`.pct_change()` da el primero; el segundo se calcula con `np.log(p / p.shift(1))`. La diferencia importa: **los logarítmicos se suman en el tiempo** ($\sum_t \ell_t = \ln(P_T/P_0)$) y los simples no; en cambio los simples se promedian correctamente **entre activos** de un portafolio y los logarítmicos no. Cada uno es aditivo en una dimensión distinta.

**`shift` es el operador de rezago.** `s.shift(1)` mueve cada valor un periodo hacia adelante, de modo que en la fila de hoy queda el dato de ayer. Es la herramienta con la que se responde a la única pregunta que importa en una serie: *¿qué sabía yo en el momento de decidir?*

**Ventanas y frecuencia.** `.rolling(n)` aplica una función a las últimas $n$ observaciones —media móvil, volatilidad móvil— y `.resample()` cambia la frecuencia, agregando de diaria a mensual. Ojo con la regla de frecuencia: `"M"` **fue eliminado** en pandas 3; la forma correcta hoy es `"ME"` (*month end*). Es un ejemplo perfecto de por qué se fija la versión de las librerías.

**El sesgo de anticipación** (*look-ahead bias*) es usar en el momento $t$ información que no existía hasta $t+1$. Es silencioso, no lanza ningún error, y produce resultados espectaculares y falsos. Es el error más caro de esta área y el último ejercicio del módulo trata exactamente de eso.

````python
# ---- Mini-lección: una serie de tiempo de principio a fin ----
import pandas as pd, numpy as np

fechas = ["2024-01-05", "2024-01-02", "2024-01-04", "2024-01-03"]   # desordenadas
precios = [105.0, 100.0, 103.0, 102.0]

# 1) De texto a indice temporal, y ORDENAR (nada funciona bien sin esto)
s = pd.Series(precios, index=pd.to_datetime(fechas)).sort_index()
print(s, "\ntipo de indice:", type(s.index).__name__, "\n")

# 2) Rendimientos: simples y logaritmicos
print("simples :", s.pct_change().round(6).tolist())
print("log     :", np.log(s / s.shift(1)).round(6).tolist())
print("suma de los log =", round(float(np.log(s / s.shift(1)).sum()), 6),
      "| log del total =", round(float(np.log(s.iloc[-1] / s.iloc[0])), 6), "<- iguales\n")

# 3) shift: lo que YA sabias al abrir el dia
print(pd.DataFrame({"precio": s, "precio_ayer": s.shift(1)}), "\n")

# 4) Ventana movil y cambio de frecuencia
diaria = pd.Series(range(1, 71), index=pd.date_range("2024-01-01", periods=70, freq="D"), dtype=float)
print("media movil 7:", diaria.rolling(7).mean().dropna().head(3).round(2).tolist())
print("a fin de mes :", diaria.resample("ME").last().tolist(), "  <- 'ME', no 'M'\n")

# 5) Alineacion automatica por fecha: lo que no coincide se vuelve NaN
a = pd.Series([1.0, 2.0, 3.0], index=pd.to_datetime(["2024-01-02", "2024-01-03", "2024-01-04"]))
b = pd.Series([10.0, 20.0], index=pd.to_datetime(["2024-01-03", "2024-01-04"]))
print("a + b:\n", a + b)
````

> 🔮 **Predice antes de correr.** ¿Cuántos elementos tendrá `a + b` en la última celda: dos (los que coinciden) o tres (la unión de las fechas)? ¿Y qué valor tendrá el que no coincide?

```python
import numpy as np

def rend_simple(precios):
    """Rendimientos simples, sin el NaN inicial."""
    return precios.pct_change().dropna()

def rend_log(precios):
    """Rendimientos logarítmicos (continuos), sin el NaN inicial."""
    return np.log(precios / precios.shift(1)).dropna()

p = pd.Series([100.0, 102.0, 101.0, 105.0, 108.0], index=pd.to_datetime(
    ["2024-01-02", "2024-01-03", "2024-01-04", "2024-01-05", "2024-01-08"]))
print("suma de log      :", round(float(rend_log(p).sum()), 6))
print("log del total    :", round(float(np.log(p.iloc[-1] / p.iloc[0])), 6), "<- idénticos")
print("suma de simples  :", round(float(rend_simple(p).sum()), 6), "<- NO es el total real")
print("total real       :", round(float(p.iloc[-1] / p.iloc[0] - 1), 6))
```

**Por qué:** No son dos formas de decir lo mismo, y elegir mal produce números que nadie detecta.

**El logarítmico es aditivo en el tiempo.** $\sum_t \ln(P_t/P_{t-1}) = \ln(P_T/P_0)$ — la suma telescópica. Por eso se usa para acumular a lo largo de una serie, y por eso la volatilidad anualizada se calcula sobre logarítmicos multiplicando por $\sqrt{252}$.

**El simple es aditivo entre activos.** El rendimiento de un portafolio es $\sum_i w_i r_i$ con rendimientos **simples**; con logarítmicos esa igualdad es falsa. Si construyes un portafolio con log-returns ponderados, el error es pequeño con movimientos chicos y crece con la volatilidad — justo cuando más te importa.

Regla práctica: **logarítmicos para acumular en el tiempo, simples para agregar entre activos.** Y nota que para movimientos pequeños casi coinciden ($\ln(1+r) \approx r$): arriba, 0.0198 contra 0.0200. Esa cercanía es lo que hace que el error pase desapercibido en datos diarios tranquilos y explote en una crisis.

Sobre el `.dropna()`: el primer rendimiento **no existe** — no hay día anterior. Rellenarlo con cero sería inventar un dato y sesgar a la baja cualquier medida de volatilidad.

</details>

```python
def media_movil(precios, ventana):
    """Media de las últimas `ventana` observaciones."""
    return precios.rolling(ventana).mean().dropna()

def volatilidad_movil(precios, ventana):
    """Desviación estándar de los rendimientos en la ventana."""
    return precios.pct_change().rolling(ventana).std().dropna()

p = pd.Series([100.0, 102.0, 101.0, 105.0, 108.0], index=pd.to_datetime(
    ["2024-01-02", "2024-01-03", "2024-01-04", "2024-01-05", "2024-01-08"]))
print(media_movil(p, 3).round(4).tolist())
print(volatilidad_movil(p, 3).round(6).tolist())
```

**Por qué:** Los `NaN` iniciales no son un defecto: con una ventana de 3, las dos primeras filas **no tienen** tres observaciones detrás. `rolling` es honesto y lo dice. Rellenar esos huecos con la media de lo que haya es exactamente el sesgo de anticipación del ejercicio 146, en versión sutil.

Nota que `rolling(n)` es una ventana **hacia atrás**: el valor en la fila $t$ resume $[t-n+1, t]$, todo pasado. Eso la vuelve segura para decidir en tiempo real. Su prima peligrosa es `.rolling(n, center=True)`, que centra la ventana y por tanto **usa datos futuros** — perfecta para describir una serie histórica en una gráfica, veneno para simular una decisión.

Y la advertencia del enunciado, que vale por todo el módulo: la volatilidad se mide sobre **rendimientos**, no sobre precios. La desviación estándar de una serie de precios crece simplemente porque el precio subió, sin que el activo se haya vuelto más riesgoso. Para anualizar la volatilidad diaria se multiplica por $\sqrt{252}$, los días hábiles del año — y eso ya supone rendimientos independientes, supuesto que casi nunca se cumple.

```python
    """Último precio observado de cada mes."""
    return precios.resample("ME").last()

                   index=pd.date_range("2024-01-01", periods=70, freq="D"))
```

**Por qué:** Dos decisiones dentro de una línea aparentemente trivial.

**Qué función agrega.** `.last()` toma el cierre; `.mean()` el promedio; `.ohlc()` da apertura, máximo, mínimo y cierre de un golpe. Para precios, el promedio mensual **no es el precio de nada** y no se puede negociar a él: encadenar rendimientos sobre promedios produce una serie más suave que la realidad y por tanto una volatilidad subestimada.

**Cuidado con el sentido de la conversión.** Bajar la frecuencia (diario → mensual) agrega y es seguro. Subirla (mensual → diario) requiere inventar, y el método por defecto que la gente usa, `.ffill()`, propaga el último valor conocido hacia adelante. Eso está bien si de verdad era lo último que sabías, y es sesgo de anticipación si el dato mensual se publicó con rezago — el PIB de un trimestre no está disponible el primer día de ese trimestre.

Y la lección de versiones: `"M"` funcionó durante una década, quedó obsoleta en pandas 2.2 y hoy lanza `ValueError`. Es la razón concreta de tener un `requirements.lock.txt`: tu código de hace tres años no falla porque estuviera mal escrito, sino porque la librería cambió debajo.

#### Ejercicio 145 — 📈 Alinear dos series con calendarios distintos

```python
def alinear(a, b):
    """DataFrame con las dos series, solo en las fechas comunes."""
    return pd.concat([a.rename("a"), b.rename("b")], axis=1, join="inner").sort_index()

a = pd.Series([1.0, 2.0, 3.0, 4.0], index=pd.to_datetime(
    ["2024-01-02", "2024-01-03", "2024-01-04", "2024-01-05"]))
b = pd.Series([10.0, 20.0, 30.0], index=pd.to_datetime(
    ["2024-01-03", "2024-01-04", "2024-01-08"]))
print(alinear(a, b))
```

**Por qué:** La alineación por fecha es la razón principal para usar un `DatetimeIndex`, y también la trampa. Cuando sumas dos series, pandas **no** empareja por posición sino por índice: si sumaras dos listas de Python de longitudes distintas tendrías un error, aquí obtienes `NaN` en las fechas que no coinciden y ningún aviso.

La elección entre `join="inner"` y `join="outer"` no es técnica, es metodológica:

- **`inner`** se queda con la intersección. Pierdes observaciones, pero cada fila que queda es real. Es lo correcto para calcular una correlación o una beta entre dos activos.
- **`outer`** conserva la unión y llena con `NaN`. Sirve para ver qué falta y por qué, antes de decidir.

Lo que casi nunca es correcto es `outer` seguido de `.fillna(0)`: un día sin cotización no es un día con precio cero, y meterlo en una correlación la destroza. Si vas a rellenar, `.ffill()` —arrastrar el último precio conocido— es defendible para precios, porque es literalmente lo último que sabías. Nunca para volúmenes ni para rendimientos.

</details>

````python
# ---------- Verificación (no la edites) ----------
_pr = pd.Series([100.0, 102.0, 101.0, 105.0, 108.0], index=pd.to_datetime(
    ["2024-01-02", "2024-01-03", "2024-01-04", "2024-01-05", "2024-01-08"]))
def _ganancia(senal, precios):
    """Rendimiento acumulado de una estrategia que sigue la señal."""
    return float((senal * precios.pct_change().fillna(0.0)).sum())

probar("Ejercicio 146", [
    (lambda: senal_mala(_pr).tolist(), [0, 1, 0, 1, 1]),
    (lambda: senal_buena(_pr).tolist(), [0, 0, 1, 0, 1]),
    (lambda: senal_buena(_pr).tolist()[1:], senal_mala(_pr).tolist()[:-1]),
    (lambda: int(senal_buena(_pr).iloc[0]), 0),
    (lambda: str(senal_buena(_pr).dtype), "int64"),
    # la version con sesgo "gana" mucho mas, y por eso es sospechosa
    (lambda: _ganancia(senal_mala(_pr), _pr) > _ganancia(senal_buena(_pr), _pr), True),
])
````

<summary>🔑 Solución del ejercicio 146 — <em>ábrela solo después de intentarlo</em></summary>

```python
def senal_mala(precios):
    r = precios.pct_change()
    return (r > 0).astype("int64")

def senal_buena(precios):
    """Posición de hoy decidida con el rendimiento de AYER."""
    r = precios.pct_change()
    return (r > 0).shift(1).fillna(False).astype("int64")

p = pd.Series([100.0, 102.0, 101.0, 105.0, 108.0], index=pd.to_datetime(
    ["2024-01-02", "2024-01-03", "2024-01-04", "2024-01-05", "2024-01-08"]))

def ganancia(senal, precios):
    return float((senal * precios.pct_change().fillna(0.0)).sum())

print("mala :", senal_mala(p).tolist())
print("buena:", senal_buena(p).tolist(), "<- la misma señal, corrida un día")
print(f"ganancia con sesgo : {ganancia(senal_mala(p), p):.4f}")
print(f"ganancia honesta   : {ganancia(senal_buena(p), p):.4f}")
```

**Por qué:** `senal_mala` no tiene ningún error de programación. Corre, no lanza excepciones, los tipos son correctos y el resultado se ve bien. Simplemente responde a una pregunta que nadie puede responder en tiempo real: *¿va a subir hoy?*

Su rendimiento en este ejemplo es casi cinco veces el de la versión honesta, y con datos reales el efecto es mucho más brutal: una estrategia que solo compra los días que suben nunca pierde, y produce curvas de capital que suben en línea recta. **Ese es el síntoma.** Un backtest que se ve demasiado bien no es una buena estrategia; es, casi siempre, una fuga de información desde el futuro.

El sesgo se cuela por más rendijas de las que parece:

| Fuente | Cómo se ve |
|---|---|
| Decidir con el dato del mismo periodo | lo de este ejercicio: falta un `shift(1)` |
| Normalizar con la media de **toda** la muestra | usas el promedio de 2024 para decidir en enero de 2024 |
| `rolling(n, center=True)` | la ventana incluye el futuro |
| Datos revisados después | el PIB que ves hoy no es el que se publicó ese día |
| Universo de acciones actual | analizar "las empresas del índice hoy" excluye a las que quebraron (*survivorship bias*) |

La defensa es una sola pregunta, aplicada a cada línea que toque datos ordenados en el tiempo: **¿esta información existía en el momento en que la estoy usando?** Si la respuesta no es un sí evidente, hay un `shift` faltante.

</details>

## Módulo 17 — Proyecto integrador

**Ejercicios 147–148** · duración aproximada: 45–90 min

**Al terminar este módulo podrás:**

- Encadenar validación, limpieza, cálculo y reporte en una sola función
- Devolver los errores como datos en lugar de dejar que tumben el proceso
- Producir un resultado reutilizable: métricas, gráfica y fuente en un diccionario

### Teoría

No hay teoría nueva. Los dos ejercicios de este módulo son de **integración**: no introducen ni una función ni una sintaxis que no hayas usado ya. Su dificultad es de otro tipo, y es la más parecida al trabajo real: si cualquiera de las piezas que escribiste en los módulos anteriores tiene un bug, aquí se manifiesta.

El patrón que vas a implementar es el de cualquier análisis serio:

```
leer  →  validar  →  limpiar  →  transformar  →  calcular  →  reportar
```

Con dos reglas de diseño que ya practicaste y que aquí se vuelven visibles:

1. **Falla temprano.** La validación va al principio, y si algo no cuadra se devuelve el error como **dato** (`{"error": ...}`) en lugar de dejar que el proceso truene a la mitad o —peor— que siga con datos corruptos.
2. **Devuelve, no imprimas.** Un reporte que devuelve un diccionario se puede guardar en JSON, mandar por correo, comparar contra el del mes pasado o probar automáticamente. Uno que imprime, no sirve para nada de eso.

Si terminas los dos ejercicios, tienes un proceso que va de un archivo en disco a un conjunto de métricas y una gráfica, sin intervención manual. Cambiarle el archivo de entrada es todo lo que hace falta para reutilizarlo.

````python
# ---- Mini-lección: cómo se ve un pipeline por dentro ----
# Este es el esqueleto genérico. Los dos ejercicios lo llenan con tus funciones.

def pipeline_generico(fuente, validar, limpiar, calcular, reportar):
    """Cada paso es una función; el pipeline solo los encadena y corta ante error."""
    datos = fuente()
    problema = validar(datos)
    if problema:
        return {"error": problema}          # falla temprano, devuelve el error como dato
    limpios = limpiar(datos)
    metricas = calcular(limpios)
    return reportar(metricas)

# Ejemplo mínimo, con funciones de juguete
resultado = pipeline_generico(
    fuente=lambda: [1, 2, None, 4],
    validar=lambda d: None if d else "sin datos",
    limpiar=lambda d: [x for x in d if x is not None],
    calcular=lambda d: {"n": len(d), "suma": sum(d), "media": sum(d) / len(d)},
    reportar=lambda m: {**m, "version": "1.0"},
)
print(resultado)

# Y el caso de error, que devuelve dato en lugar de explotar
print(pipeline_generico(lambda: [], lambda d: None if d else "sin datos",
                        lambda d: d, lambda d: {}, lambda m: m))
````

```python
def reporte(df):
    """Reporte económico a partir de un DataFrame de países.

    Espera las columnas: pais, pib (miles de millones USD),
    poblacion (millones) e inflacion (porcentaje anual).
    Devuelve un diccionario de métricas, o {"error": mensaje}
    si la estructura de entrada no es válida.
    """
    requeridas = {"pais", "pib", "poblacion", "inflacion"}
    faltantes = requeridas - set(df.columns)
    if faltantes:
        return {"error": f"faltan columnas: {sorted(faltantes)}"}

    # Limpieza: fuera los renglones con datos numéricos faltantes
    d = df.dropna(subset=["pib", "poblacion", "inflacion"]).copy()
    if len(d) == 0:
        return {"error": "no quedaron renglones válidos tras limpiar"}

    # Columna calculada (vectorizada)
    d["pib_per_capita"] = d["pib"] / d["poblacion"] * 1000

    pibpc = d["pib_per_capita"].tolist()
    infl = d["inflacion"].tolist()

    g = gini(pibpc)
    r = correlacion(pibpc, infl)

    return {
        "n": int(len(d)),
        "pib_total": float(d["pib"].sum()),
        "pib_per_capita_promedio": float(d["pib_per_capita"].mean()),
        "pais_max_inflacion": str(d.loc[d["inflacion"].idxmax(), "pais"]),
        "gini_pib_per_capita": round(g, 4) if g is not None else None,
        "correlacion_pibpc_inflacion": round(r, 4) if r is not None else None,
    }

import json

demo = pd.DataFrame({
    "pais": ["MX", "BR", "CL", "AR"],
    "pib": [1790.0, 2170.0, 340.0, 640.0],
    "poblacion": [129.0, 216.0, 19.6, 46.0],
    "inflacion": [4.7, 4.6, 3.9, 211.4],
})
print(json.dumps(reporte(demo), indent=2, ensure_ascii=False))
print(reporte(pd.DataFrame({"pais": ["MX"], "pib": [1790.0]})))   # caso inválido
```

**Por qué:** Este es el esqueleto de cualquier *pipeline* de análisis serio: **validar → limpiar → transformar → agregar → reportar**, con la validación al principio y errores devueltos como datos en lugar de excepciones que tumban el proceso. Devolver un diccionario (en vez de imprimir) permite encadenar el reporte a un archivo, un correo automatizado o una prueba unitaria. Si llegaste aquí con los ejercicios en verde, hiciste algo que muy poca gente que "empieza un curso de Python" termina. Vale la pena ser explícito sobre **qué sabes hacer ahora**, porque es más de lo que parece:

- expresar un cálculo cualquiera en código, empaquetarlo en una función documentada y probarla con casos frontera;
- traducir un conjunto de reglas (una tarifa, un criterio de elegibilidad, un artículo de un lineamiento) a lógica condicional con prioridades explícitas, y detectar cuándo esa lógica tiene código muerto;

- leer cualquier `traceback` y escribir código que sobrevive a datos sucios sin inventar valores;

- **leer datos que no escribiste tú**: archivos de texto, CSV con las convenciones de quien los hizo, JSON anidado de una API, y producir una gráfica guardada en disco;
- implementar estadísticos desde su definición matemática —correlación, Gini, MCO, prueba t, intervalo de confianza— **validarlos contra la librería** y saber qué **no** significan;
- encadenar todo eso en un proceso que va de un archivo a un reporte sin intervención manual.

### Hacia dónde seguir

Tres caminos, según a dónde vayas:

**Ciencia de datos aplicada.** Ya sabes leer un CSV y consumir una API; el siguiente paso es hacerlo con fuentes de verdad — microdatos del INEGI, series de Banxico (SIE), el catálogo del Banco Mundial, precios con `yfinance` — y visualización más seria con `seaborn` o `plotly`. El cuello de botella ya no es Python: es saber qué preguntarle a los datos.
